## 1. Why Financial Statements Matter

Financial statements are the **universal language of business**. Every publicly traded company in the world distils its economic activity into the same small set of reports, enabling investors, creditors, regulators, and managers to evaluate performance on a level playing field.

Three distinct user groups depend on these reports for fundamentally different reasons:

| User | Primary Question | Key Statement |
|------|-----------------|---------------|
| **Equity investor** | Is the company profitable and growing? | Income statement |
| **Creditor / lender** | Can the company repay its debts? | Balance sheet & cash flow |
| **Manager** | Where is cash being generated and consumed? | Cash flow statement |

> **Key Concept:** Financial statements reduce *information asymmetry* — the gap between what management knows about the company and what outsiders can observe. Without standardised reporting, capital markets could not function.

### The reporting ecosystem

Financial statements do not exist in isolation. They sit inside a broader ecosystem:

1. **Accounting standards** (IFRS or US GAAP) dictate *how* transactions are recorded.
2. **External auditors** provide assurance that the statements are materially correct.
3. **Regulators** (e.g. SEC, FCA) mandate *what* must be disclosed and *when*.
4. **Analysts** interpret the statements to form investment recommendations.

> **CFA Exam Tip:** The CFA curriculum emphasises that financial statements are *general purpose* — they aim to serve a broad audience, not any single user. Analysts must therefore extract the specific information they need through ratio analysis and adjustments (covered in later notebooks in this series).

### The role of financial statements in capital allocation

Financial statements serve a critical function beyond informing individual investment decisions — they enable **efficient capital allocation** across the economy. When statements are transparent and reliable:

* Capital flows toward companies that generate the highest risk-adjusted returns.
* Poorly managed companies face higher costs of capital, incentivising improvement or exit.
* Creditors can accurately price the risk of lending, keeping credit markets functioning.

When financial reporting fails — as in the Enron and WorldCom scandals of the early 2000s — capital is misallocated. Investors lose money on fraudulent companies that should never have attracted capital, while legitimate companies are starved of funding as trust in financial reporting erodes.

> **Common Mistake:** Students often view financial statements as purely backward-looking documents. In fact, investors use historical financial data primarily to *forecast* future performance. The past matters only insofar as it informs expectations about future cash flows, growth, and risk.

## 2. The Accounting Equation

Every financial statement rests on one identity — the **accounting equation**:

$$
\text{Assets} = \text{Liabilities} + \text{Equity}
$$

This is not an approximation; it holds **exactly** after every single transaction. To see why, consider what a company *is* from a financial perspective:

* **Assets** — resources the company controls (cash, inventory, equipment, patents).
* **Liabilities** — obligations to *others* (loans, accounts payable, deferred revenue).
* **Equity** — the residual claim belonging to *owners* after all liabilities are settled.

If a company owns \$100 of assets and owes \$60 to creditors, the owners' claim is exactly \$40. That is the accounting equation in action.

### Why double-entry bookkeeping preserves the equation

The brilliance of double-entry bookkeeping (first codified by Luca Pacioli in 1494) is that **every transaction touches at least two accounts**, keeping the equation in balance:

| Transaction | Debit (increase) | Credit (increase) |
|------------|-------------------|---------------------|
| Company receives \$10,000 cash from issuing shares | Cash (Asset) +\$10,000 | Share Capital (Equity) +\$10,000 |
| Company borrows \$5,000 from a bank | Cash (Asset) +\$5,000 | Bank Loan (Liability) +\$5,000 |
| Company buys inventory for \$3,000 cash | Inventory (Asset) +\$3,000 | Cash (Asset) −\$3,000 |
| Company earns \$4,000 revenue on credit | Accounts Receivable (Asset) +\$4,000 | Revenue (Equity via Retained Earnings) +\$4,000 |

> **Key Concept:** *Debits* and *credits* are not "good" or "bad" — they are simply the left and right sides of each journal entry. Assets increase with debits; liabilities and equity increase with credits. The total debits always equal total credits.

### Expanding the equation

For analysis, we often expand equity into its components:

$$
\text{Assets} = \text{Liabilities} + \text{Contributed Capital} + \text{Retained Earnings}
$$

where

$$
\text{Retained Earnings} = \text{Beginning RE} + \text{Net Income} - \text{Dividends}
$$

and

$$
\text{Net Income} = \text{Revenue} - \text{Expenses}
$$

This chain links all three financial statements: the income statement feeds retained earnings, retained earnings feed the balance sheet, and changes in the balance sheet feed the cash flow statement.

> **CFA Exam Tip:** The expanded accounting equation is the conceptual backbone of the CFA L1 FSA curriculum. If you ever get lost in the details, return to A = L + E.

### Why the accounting equation always holds

This is not a coincidence or an approximation — the equation is *tautologically true by construction*. Equity is **defined** as the residual: Assets minus Liabilities. It is impossible for the equation to be "out of balance" any more than it is possible for your height minus your inseam to not equal your torso length — one is defined as the difference of the other two.

Double-entry bookkeeping is simply the *mechanism* that preserves this identity at the transaction level, ensuring that each individual event keeps the books in balance rather than waiting until year-end to compute the residual.

> **CFA Exam Tip:** When working exam problems, you can always use A = L + E to solve for any missing value. If you know total assets and total liabilities, equity is determined. This is especially useful in questions that provide partial balance sheet information and ask you to compute a missing line item.

## 4. Accrual vs Cash Accounting

One of the most important distinctions in financial reporting is between **accrual accounting** and **cash accounting**.

### Cash accounting

Under cash accounting, revenue is recorded when cash is *received* and expenses are recorded when cash is *paid*. This is intuitive — it mirrors your personal bank account.

**Problem:** suppose a consulting firm completes a \$50,000 project in December but the client doesn't pay until January. Under cash accounting, December's statements show zero revenue for work that was fully performed. This distorts economic reality.

### Accrual accounting

Under accrual accounting, revenue is recorded when it is *earned* (i.e. the goods/services have been delivered), and expenses are recorded when they are *incurred* (i.e. the economic benefit has been consumed), **regardless of when cash changes hands**.

| Event | Cash Accounting | Accrual Accounting |
|-------|----------------|-------------------|
| Deliver goods in Dec, receive payment in Jan | Revenue in Jan | Revenue in Dec |
| Receive rent payment in Dec for Jan occupancy | Revenue in Dec | Revenue in Jan |
| Use electricity in Dec, pay bill in Jan | Expense in Jan | Expense in Dec |

> **Key Concept:** All major accounting standards (IFRS and US GAAP) require **accrual accounting** for published financial statements. The gap between accrual profits and actual cash flows is one of the central themes of financial statement analysis.

### Why accrual accounting dominates

Accrual accounting better matches **economic substance** with **reporting periods**:

* **Revenue** is recognised when the *performance obligation* is satisfied — the value has been created.
* **Expenses** are recognised in the same period as the revenue they helped generate (*matching principle*).

This creates a more faithful picture of profitability, but it also introduces *estimates and judgments* — allowance for bad debts, depreciation schedules, warranty provisions — which is precisely why analysts must scrutinise financial statements rather than take them at face value.

> **Common Mistake:** Students sometimes assume that higher net income means more cash. It does not. A company can report record profits while running out of cash (e.g. if it extends generous credit terms and customers are slow to pay). The cash flow statement exists precisely to reconcile this gap.

### Worked example — accrual vs cash

Let us build a minimal example. A company in its first year has the following transactions:

| Month | Event | Cash In/Out | Revenue (Accrual) | Expense (Accrual) |
|-------|-------|-------------|-------------------|-------------------|
| Jan | Sell \$100 of goods on credit | — | \$100 | — |
| Feb | Pay \$40 for raw materials used in Jan | −\$40 | — | \$40 (Jan) |
| Mar | Collect the \$100 receivable | +\$100 | — | — |

**Cash basis:** Q1 net cash = +\$100 − \$40 = +\$60 (but all in March!)

**Accrual basis:** Q1 net income = \$100 − \$40 = \$60 (\$60 in January, \$0 in Feb/Mar)

Both methods arrive at the same total, but accrual accounting correctly assigns the profit to January, when the economic value was created.

### When does cash accounting make sense?

Despite its limitations for large companies, cash accounting is perfectly adequate for:

* **Small businesses** — a freelancer or sole proprietor with simple transactions benefits from the simplicity of cash accounting.
* **Personal finance** — your bank account *is* a cash accounting system, and it works well for managing household finances.
* **Tax planning** — in some jurisdictions, small businesses may use cash-basis accounting for tax purposes, allowing them to defer revenue by delaying invoicing.

The key distinction is between *internal* decision-making (where cash accounting may suffice) and *external* reporting to investors and regulators (where accrual accounting is mandatory).

> **Key Concept:** The gap between accrual earnings and cash flow is called **accruals**. This concept is so important that an entire later notebook in this series (Earnings Quality & Accrual Analysis) is devoted to analysing what accruals reveal about a company's financial health and the reliability of its reported earnings.

```python
# ── Accrual vs Cash: monthly comparison ──

months = np.array(['Jan', 'Feb', 'Mar'])

# Cash accounting
cash_in   = np.array([  0,   0, 100])   # collect receivable in Mar
cash_out  = np.array([  0, -40,   0])   # pay supplier in Feb
net_cash  = cash_in + cash_out

# Accrual accounting
accrual_revenue = np.array([100,  0,  0])   # earned in Jan
accrual_expense = np.array([ 40,  0,  0])   # matched to Jan
net_income      = accrual_revenue - accrual_expense

fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharey=True)

# Cash basis
x = np.arange(len(months))
axes[0].bar(x - 0.15, cash_in, 0.3, label='Cash In', color=PRIMARY)
axes[0].bar(x + 0.15, cash_out, 0.3, label='Cash Out', color=SECONDARY)
axes[0].plot(x, net_cash, 'o--', color='black', label='Net Cash')
axes[0].set_xticks(x)
axes[0].set_xticklabels(months)
axes[0].set_title('Cash Accounting')
axes[0].set_ylabel('Amount ($)')
axes[0].legend()
axes[0].axhline(0, color='grey', linewidth=0.5)

# Accrual basis
axes[1].bar(x - 0.15, accrual_revenue, 0.3, label='Revenue', color=PRIMARY)
axes[1].bar(x + 0.15, -accrual_expense, 0.3, label='Expense', color=SECONDARY)
axes[1].plot(x, net_income, 'o--', color='black', label='Net Income')
axes[1].set_xticks(x)
axes[1].set_xticklabels(months)
axes[1].set_title('Accrual Accounting')
axes[1].legend()
axes[1].axhline(0, color='grey', linewidth=0.5)

fig.suptitle('Cash vs Accrual Accounting — Same Transactions, Different Timing', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()

print(f"Cash basis — cumulative net cash by end of Q1:  ${net_cash.sum():.0f}")
print(f"Accrual basis — cumulative net income by end of Q1: ${net_income.sum():.0f}")
print("Both totals are identical; the difference is in *when* the profit appears.")
```

The chart above makes the key point vivid: under cash accounting the profit appears entirely in March (when cash arrives), while under accrual accounting the profit is correctly assigned to January (when the sale occurred). Over longer periods the cumulative totals converge, but *within* any given period the two methods can diverge sharply.

This divergence has real consequences for decision-making. Consider a bank evaluating ExampleCo for a loan in February:
- Under **cash accounting**, February shows zero revenue and a \$40 cash outflow — the company looks like it is haemorrhaging cash.
- Under **accrual accounting**, the bank sees that January generated \$60 of profit and the February cash outflow is simply a timing difference.

The accrual picture is more faithful to economic reality, which is precisely why regulators mandate it for published financial statements.

> **Common Mistake:** Students sometimes confuse the *timing* of revenue recognition with the *amount*. Accrual accounting does not create revenue out of thin air — it records the same total revenue as cash accounting, just in different periods. The lifetime totals always converge.

## 5. The Three Core Financial Statements

Every publicly listed company must publish three primary statements each reporting period:

### 5.1 The Income Statement (Statement of Profit or Loss)

The income statement measures **profitability over a period** (quarter or year). It answers: *"How much value did the company create (or destroy) during this period?"*

**Structure (multi-step format):**

$$
\begin{aligned}
&\text{Revenue} \\
-\;& \text{Cost of Goods Sold (COGS)} \\
=\;& \textbf{Gross Profit} \\
-\;& \text{Operating Expenses (SG\&A, R\&D, D\&A)} \\
=\;& \textbf{Operating Income (EBIT)} \\
-\;& \text{Interest Expense} \\
+\;& \text{Interest / Other Income} \\
=\;& \textbf{Earnings Before Tax (EBT)} \\
-\;& \text{Income Tax Expense} \\
=\;& \textbf{Net Income}
\end{aligned}
$$

> **Key Concept:** Each line on the income statement represents a **margin** — gross margin, operating margin, net margin. Tracking how these margins change over time reveals whether a company is becoming more or less efficient.

### 5.2 The Balance Sheet (Statement of Financial Position)

The balance sheet is a **snapshot at a point in time**. It answers: *"What does the company own, what does it owe, and what is left for shareholders?"*

It is organised around the accounting equation:

| Assets | = | Liabilities | + | Equity |
|--------|---|-------------|---|--------|
| Current assets (cash, receivables, inventory) | | Current liabilities (payables, short-term debt) | | Share capital |
| Non-current assets (PP&E, intangibles, investments) | | Non-current liabilities (long-term debt, pensions) | | Retained earnings |

> **CFA Exam Tip:** The balance sheet is a *stock* variable (measured at an instant), while the income statement and cash flow statement are *flow* variables (measured over a period). Mixing stocks and flows in ratios (e.g. ROA = NI / Average Assets) requires averaging the balance sheet values.

### 5.3 The Cash Flow Statement

The cash flow statement tracks **actual cash movements** over a period, classified into three activities:

1. **Operating (CFO)** — cash generated by the core business (collections from customers minus payments to suppliers and employees).
2. **Investing (CFI)** — cash spent on or received from long-term assets (buying equipment, selling a subsidiary).
3. **Financing (CFF)** — cash from or to capital providers (issuing shares, repaying debt, paying dividends).

$$
\Delta \text{Cash} = \text{CFO} + \text{CFI} + \text{CFF}
$$

> **Key Concept:** The cash flow statement acts as a *bridge* between the income statement (accrual-based profits) and the balance sheet (which includes cash). It is often considered the hardest statement to manipulate, making it a favourite tool for analysts assessing earnings quality.

### How the three statements relate to each other

Think of the three statements as three different cameras filming the same company:

* The **income statement** is a *movie* — it shows what happened over the entire year.
* The **balance sheet** is a *photograph* — it freezes the company at December 31st.
* The **cash flow statement** is a *bank statement* — it tracks every dollar in and out.

None of these views alone is sufficient. A company can be profitable (income statement) but cash-poor (cash flow statement). It can have strong cash flow but be insolvent (liabilities exceed assets on the balance sheet). Only by reading all three together can an analyst form a complete picture.

### The "fourth statement" — Statement of Changes in Equity

Although the CFA curriculum focuses on three statements, companies also publish a **Statement of Changes in Equity** that reconciles beginning and ending equity balances. It captures:
* Net income (from the income statement)
* Dividends declared
* Share issuances and buybacks
* Other comprehensive income (OCI) items

> **Key Concept:** The Statement of Changes in Equity is the formal bridge between the income statement and the balance sheet. It shows exactly how the equity section of the balance sheet changed from the start to the end of the period.

### Direct vs indirect method for operating cash flow

There are two methods for presenting CFO:

* **Direct method:** Lists actual cash receipts (from customers) and payments (to suppliers, employees, etc.). More intuitive but rarely used in practice because it requires tracking every cash transaction separately.

* **Indirect method:** Starts with net income and *adjusts* for non-cash items (depreciation, gains/losses) and working capital changes (receivables, inventory, payables). Used by >95% of companies because it reconciles easily with the income statement.

Both methods produce **identical CFO** — the difference is purely in presentation. The indirect method is covered in depth in the Cash Flow Statement Analysis notebook.

> **CFA Exam Tip:** You must be comfortable with both methods. The exam may ask you to convert between them or to explain why CFO differs from net income. The key non-cash adjustments to know are: add back depreciation/amortisation, subtract gains on asset sales, and adjust for working capital changes (increase in receivables = subtract, increase in payables = add).

### Which statement matters most?

Different users prioritise different statements:

| Stakeholder | Primary Focus | Reasoning |
|-------------|--------------|-----------|
| Equity analyst | Income statement → cash flow | Earnings growth drives stock prices; cash flow validates earnings quality |
| Credit analyst | Cash flow → balance sheet | Debt service requires cash; balance sheet shows collateral and leverage |
| Venture capitalist | Cash flow statement | Cash burn rate determines runway — profitability is secondary for early-stage companies |
| Regulator | All three equally | Regulatory compliance requires a complete, consistent set of statements |

## 6. Revenue and Expense Recognition

### Revenue recognition — IFRS 15 / ASC 606

Both IFRS and US GAAP converged on a single **five-step model** for revenue recognition:

| Step | Description | Example |
|------|-------------|---------|
| 1. Identify the contract | Agreement with a customer that creates enforceable rights/obligations | Signed software licence agreement |
| 2. Identify performance obligations | Distinct goods/services promised | Licence + 1 year of support = 2 obligations |
| 3. Determine transaction price | Amount the company expects to receive | \$120,000 |
| 4. Allocate price to obligations | Based on standalone selling prices | Licence \$90,000 / Support \$30,000 |
| 5. Recognise revenue when obligations are satisfied | Over time or at a point in time | Licence at delivery / Support over 12 months |

> **CFA Exam Tip:** The distinction between "over time" and "at a point in time" recognition is a frequent exam topic. Construction contracts (percentage-of-completion) recognise over time; retail sales recognise at a point in time.

### Expense recognition — the matching principle

Expenses are recognised in the same period as the revenue they help generate:

* **Direct matching:** COGS is matched to the specific revenue from selling those goods.
* **Period costs:** SG&A expenses are expensed in the period incurred (they benefit the current period generally).
* **Systematic allocation:** Depreciation spreads the cost of a long-lived asset over its useful life.

> **Common Mistake:** Capitalising vs expensing is one of the most impactful accounting choices. If a company capitalises a cost (records it as an asset), current-period expenses are *lower* and profits are *higher*. This doesn't change total lifetime expense — it only shifts the timing. Analysts must be alert to aggressive capitalisation policies.

### Why revenue recognition matters so much

Revenue is the single most important line on the income statement — all margins and most ratios derive from it. If revenue is overstated, the entire income statement is distorted.

Historically, revenue recognition was the **most common source of financial fraud**. Before the convergence to IFRS 15 / ASC 606, different industries used different standards (software revenue had its own rules, construction had percentage-of-completion, etc.), creating inconsistency and opportunities for manipulation.

The five-step model was designed to create a single, principles-based framework applicable across all industries. Its key innovations include:

1. **Performance obligation focus** — revenue is not about cash receipt or contract signing; it is about *delivering* goods or services.
2. **Variable consideration** — if the transaction price is uncertain (e.g., includes rebates, returns, or performance bonuses), the company must estimate the most likely amount.
3. **Contract modifications** — changes to scope or price are handled systematically rather than ad hoc.

> **Key Concept:** Under the five-step model, a company cannot recognise revenue simply because it signed a contract or sent an invoice. Revenue recognition requires **satisfaction of a performance obligation** — the customer must have received the promised good or service. This principle prevents companies from "pulling forward" revenue from future periods.

### Expense recognition in more depth

The three methods of expense recognition reflect fundamentally different relationships between costs and revenue:

| Method | Matching Logic | Example |
|--------|---------------|---------|
| **Direct matching** | Cost directly linked to specific revenue | Raw materials used in a product that was sold |
| **Period allocation** | Cost benefits the current period generally | CEO salary, office rent |
| **Systematic allocation** | Cost spread over the asset's useful life | Depreciation of a factory over 20 years |

> **CFA Exam Tip:** The exam often tests the distinction between costs that should be **capitalised** (recorded as an asset and depreciated) versus **expensed** (charged to the income statement immediately). The general rule: if the cost provides economic benefit beyond one year, capitalise it; if it is consumed within the period, expense it. Aggressive companies push the boundary by capitalising costs that arguably should be expensed — a common earnings manipulation technique (explored in the Earnings Quality notebook).

## 7. Building Statements from Transactions

Let us now build all three financial statements from scratch using a set of realistic transactions for **ExampleCo**, a small manufacturing firm in its first year of operations.

### Transaction ledger

| # | Description | Debit Account | Amount | Credit Account | Amount |
|---|-------------|---------------|--------|----------------|--------|
| T1 | Issue 10,000 shares at \$10 each | Cash | 100,000 | Share Capital | 100,000 |
| T2 | Borrow \$50,000 from bank (2-year loan) | Cash | 50,000 | Bank Loan | 50,000 |
| T3 | Purchase equipment for \$60,000 cash | Equipment | 60,000 | Cash | 60,000 |
| T4 | Purchase raw materials on credit | Inventory | 25,000 | Accounts Payable | 25,000 |
| T5 | Sell goods for \$80,000 (\$50,000 cash, \$30,000 on credit) | Cash / A/R | 80,000 | Revenue | 80,000 |
| T6 | COGS for goods sold | COGS | 18,000 | Inventory | 18,000 |
| T7 | Pay operating expenses (salaries, rent) | OpEx | 20,000 | Cash | 20,000 |
| T8 | Pay interest on bank loan (5% annual) | Interest Expense | 2,500 | Cash | 2,500 |
| T9 | Record depreciation (equipment, 10-year SL) | Depreciation | 6,000 | Accum. Depreciation | 6,000 |
| T10 | Pay \$15,000 to supplier (partial A/P) | Accounts Payable | 15,000 | Cash | 15,000 |
| T11 | Income tax expense (25% rate) | Tax Expense | 8,375 | Tax Payable | 8,375 |
| T12 | Declare and pay \$5,000 dividend | Dividends | 5,000 | Cash | 5,000 |

> **Key Concept:** Notice that every transaction has equal debits and credits — the accounting equation is preserved at each step. Non-cash transactions (T6 depreciation, T9 depreciation) affect the income statement and balance sheet but do *not* appear in the cash flow statement's operating section under the direct method.

### How to read the transaction ledger

Each row in the ledger records a single economic event using double-entry bookkeeping:

* The **debit account** is the account that receives value (assets increase, expenses increase).
* The **credit account** is the account that gives value (liabilities increase, equity increases, revenue increases).
* The amounts always balance — every debit has an equal credit.

As you work through the code below, trace each transaction mentally:
- Does it affect only the balance sheet (e.g., buying equipment with cash)?
- Does it affect both the income statement and balance sheet (e.g., recording revenue)?
- Does it involve cash (affects the cash flow statement) or is it non-cash (e.g., depreciation)?

> **Common Mistake:** Transaction T9 (depreciation) is a common source of confusion. No cash moves — it is purely an accounting adjustment that reduces the asset's carrying value and records an expense. But this non-cash expense is critically important: it ensures the income statement reflects the economic cost of using the equipment, even though the cash was spent when the equipment was purchased (T3).

### Simplifications in this example

ExampleCo's first year is intentionally simplified to focus on the mechanics:
* **No accounts receivable collections** — in reality, most receivables would be collected within 30-90 days.
* **Single depreciation method** — straight-line on a single asset. Real companies have dozens of assets with different useful lives.
* **Statutory = effective tax rate** — no deferred taxes, no tax credits, no jurisdictional differences.
* **Cash dividends only** — no stock dividends, no share repurchases.

Despite these simplifications, the exercise demonstrates the complete cycle: transactions → journal entries → trial balance → financial statements. Every real-world financial statement, no matter how complex, is built from exactly this same process — just with thousands more transactions.

```python
# ── Define ExampleCo's first-year transactions ──

# We represent each transaction as its effects on major accounts.
# This mirrors how a general ledger aggregates journal entries.

# Asset accounts
cash            = 0
accounts_recv   = 0
inventory       = 0
equipment       = 0
accum_depr      = 0   # contra-asset (negative effect on net assets)

# Liability accounts
accounts_pay    = 0
bank_loan       = 0
tax_payable     = 0

# Equity accounts
share_capital   = 0
retained_earn   = 0

# Income statement accumulators
revenue         = 0
cogs            = 0
opex            = 0
interest_exp    = 0
depreciation    = 0
tax_expense     = 0
dividends       = 0

# ── Process transactions ──

# T1: Issue shares
cash += 100_000;  share_capital += 100_000

# T2: Bank loan
cash += 50_000;   bank_loan += 50_000

# T3: Purchase equipment
equipment += 60_000;  cash -= 60_000

# T4: Purchase inventory on credit
inventory += 25_000;  accounts_pay += 25_000

# T5: Sell goods ($50k cash + $30k on credit)
cash += 50_000;  accounts_recv += 30_000;  revenue += 80_000

# T6: Cost of goods sold
cogs += 18_000;  inventory -= 18_000

# T7: Pay operating expenses
opex += 20_000;  cash -= 20_000

# T8: Pay interest
interest_exp += 2_500;  cash -= 2_500

# T9: Depreciation (non-cash)
depreciation += 6_000;  accum_depr += 6_000

# T10: Partial payment to supplier
accounts_pay -= 15_000;  cash -= 15_000

# T11: Income tax (computed below — 25% of EBT)
ebt = revenue - cogs - opex - depreciation - interest_exp
tax_expense = ebt * 0.25
tax_payable += tax_expense

# T12: Dividends
dividends += 5_000;  cash -= 5_000

# Compute net income and retained earnings
net_income = revenue - cogs - opex - depreciation - interest_exp - tax_expense
retained_earn = net_income - dividends

# ── Verify accounting equation ──
total_assets = cash + accounts_recv + inventory + equipment - accum_depr
total_liabilities = accounts_pay + bank_loan + tax_payable
total_equity = share_capital + retained_earn

print("=" * 55)
print("ACCOUNTING EQUATION VERIFICATION")
print("=" * 55)
print(f"Total Assets:                ${total_assets:>12,.2f}")
print(f"Total Liabilities + Equity:  ${total_liabilities + total_equity:>12,.2f}")
print(f"Balanced:                    {np.isclose(total_assets, total_liabilities + total_equity)}")
```

The accounting equation balances perfectly — as it must. Every transaction we processed maintained the A = L + E identity through equal debits and credits.

Let us now present the three formal financial statements that would be published for ExampleCo. Notice how each statement draws from the same underlying transaction data but presents it in a different format for a different analytical purpose:

- The **income statement** shows whether the company was profitable during the year.
- The **balance sheet** shows what the company owns and owes at year-end.
- The **cash flow statement** shows where cash came from and where it went.

### Reading the Income Statement

The income statement follows the multi-step format introduced in the outline. Each line deducts a category of cost from revenue, producing intermediate profit measures. Pay attention to how the \$80,000 of revenue is progressively reduced by operating costs, interest, and taxes.

```python
# ── Income Statement ──

print("=" * 55)
print("ExampleCo — INCOME STATEMENT")
print("For the Year Ended 31 December Year 1")
print("=" * 55)
print(f"Revenue                       ${revenue:>12,.2f}")
print(f"Cost of Goods Sold            $({cogs:>11,.2f})")
print(f"                              {'─' * 13}")
print(f"Gross Profit                  ${revenue - cogs:>12,.2f}")
print(f"Operating Expenses            $({opex:>11,.2f})")
print(f"Depreciation & Amortisation   $({depreciation:>11,.2f})")
print(f"                              {'─' * 13}")
print(f"Operating Income (EBIT)       ${revenue - cogs - opex - depreciation:>12,.2f}")
print(f"Interest Expense              $({interest_exp:>11,.2f})")
print(f"                              {'─' * 13}")
print(f"Earnings Before Tax (EBT)     ${ebt:>12,.2f}")
print(f"Income Tax Expense (25%)      $({tax_expense:>11,.2f})")
print(f"                              {'─' * 13}")
print(f"Net Income                    ${net_income:>12,.2f}")
print("=" * 55)
```

ExampleCo earned \$25,125 in net income on \$80,000 of revenue — a net margin of approximately 31%. The multi-step format reveals that:

* **Gross profit margin** = (\$80,000 − \$18,000) / \$80,000 = **77.5%** — high, suggesting low direct production costs.
* **Operating margin** = \$36,000 / \$80,000 = **45%** — still healthy after operating expenses and depreciation.
* **Net margin** = \$25,125 / \$80,000 = **31.4%** — reduced by interest expense and taxes.

These are unrealistically high margins for a manufacturer (designed for pedagogical clarity), but the *structure* — revenue progressively reduced by layers of cost — is exactly how real income statements work.

> **CFA Exam Tip:** Always examine the progression from gross margin to operating margin to net margin. If gross margin is stable but operating margin is declining, the problem is in operating expenses (SG&A, R&D), not in production costs.

```python
# ── Balance Sheet ──

print("=" * 55)
print("ExampleCo — BALANCE SHEET")
print("As at 31 December Year 1")
print("=" * 55)
print("ASSETS")
print("Current Assets")
print(f"  Cash                        ${cash:>12,.2f}")
print(f"  Accounts Receivable         ${accounts_recv:>12,.2f}")
print(f"  Inventory                   ${inventory:>12,.2f}")
current_assets = cash + accounts_recv + inventory
print(f"  Total Current Assets        ${current_assets:>12,.2f}")
print()
print("Non-Current Assets")
print(f"  Equipment                   ${equipment:>12,.2f}")
print(f"  Less: Accum. Depreciation   $({accum_depr:>11,.2f})")
net_ppe = equipment - accum_depr
print(f"  Net PP&E                    ${net_ppe:>12,.2f}")
print(f"                              {'─' * 13}")
print(f"TOTAL ASSETS                  ${total_assets:>12,.2f}")
print()
print("LIABILITIES")
print("Current Liabilities")
print(f"  Accounts Payable            ${accounts_pay:>12,.2f}")
print(f"  Tax Payable                 ${tax_payable:>12,.2f}")
current_liab = accounts_pay + tax_payable
print(f"  Total Current Liabilities   ${current_liab:>12,.2f}")
print()
print("Non-Current Liabilities")
print(f"  Bank Loan                   ${bank_loan:>12,.2f}")
print(f"                              {'─' * 13}")
print(f"TOTAL LIABILITIES             ${total_liabilities:>12,.2f}")
print()
print("EQUITY")
print(f"  Share Capital               ${share_capital:>12,.2f}")
print(f"  Retained Earnings           ${retained_earn:>12,.2f}")
print(f"                              {'─' * 13}")
print(f"TOTAL EQUITY                  ${total_equity:>12,.2f}")
print(f"                              {'─' * 13}")
print(f"TOTAL LIABILITIES + EQUITY    ${total_liabilities + total_equity:>12,.2f}")
print("=" * 55)
```

### Reading the Balance Sheet

The balance sheet organises ExampleCo's financial position into current and non-current categories. Several features are worth noting:

* **Cash (\$97,500)** is the largest current asset — unsurprising for a first-year company that just raised equity and debt capital.
* **Accounts receivable (\$30,000)** represents the credit sales from transaction T5 that haven't been collected yet. This will convert to cash in the next period.
* **Inventory (\$7,000)** is the unsold portion of the \$25,000 purchased in T4 (\$25,000 − \$18,000 COGS = \$7,000).
* **Net PP&E (\$54,000)** is the \$60,000 equipment less \$6,000 of accumulated depreciation from one year of straight-line depreciation.

On the right side, the \$10,000 accounts payable represents the unpaid portion of the supplier invoice (\$25,000 − \$15,000 paid = \$10,000).

### Reading the Cash Flow Statement

The cash flow statement uses the **direct method** — it tracks actual cash receipts and payments. This is the more intuitive format but is less commonly used in practice (the indirect method, covered in the Cash Flow Statement Analysis notebook, is far more prevalent).

Key observation: operating cash flow (\$12,500) is substantially *lower* than net income (\$25,125). The gap is explained by:
- \$30,000 of revenue not yet collected (accounts receivable)
- \$15,000 paid to suppliers (accounts payable reduction)
- No income tax was paid in cash this year (the \$8,375 is accrued as tax payable)

This is a perfect illustration of why accrual profits and cash flow diverge — and why both perspectives are needed.

```python
# ── Cash Flow Statement (Direct Method) ──

# Operating activities (direct method): actual cash receipts and payments
cfo_collections   =  50_000          # T5 cash portion
cfo_supplier_pay  = -15_000          # T10
cfo_opex_pay      = -20_000          # T7
cfo_interest_pay  =  -2_500          # T8
# Note: no tax was paid in cash this year (tax is payable next year)
cfo_total = cfo_collections + cfo_supplier_pay + cfo_opex_pay + cfo_interest_pay

# Investing activities
cfi_equipment = -60_000              # T3
cfi_total = cfi_equipment

# Financing activities
cff_shares   = 100_000              # T1
cff_loan     =  50_000              # T2
cff_dividend =  -5_000              # T12
cff_total = cff_shares + cff_loan + cff_dividend

net_change_cash = cfo_total + cfi_total + cff_total

print("=" * 55)
print("ExampleCo — CASH FLOW STATEMENT (Direct Method)")
print("For the Year Ended 31 December Year 1")
print("=" * 55)
print("OPERATING ACTIVITIES")
print(f"  Cash collected from customers  ${cfo_collections:>10,.2f}")
print(f"  Cash paid to suppliers         $({-cfo_supplier_pay:>9,.2f})")
print(f"  Cash paid for operating exp.   $({-cfo_opex_pay:>9,.2f})")
print(f"  Interest paid                  $({-cfo_interest_pay:>9,.2f})")
print(f"  Net Cash from Operations       ${cfo_total:>10,.2f}")
print()
print("INVESTING ACTIVITIES")
print(f"  Purchase of equipment          $({-cfi_equipment:>9,.2f})")
print(f"  Net Cash from Investing        ${cfi_total:>10,.2f}")
print()
print("FINANCING ACTIVITIES")
print(f"  Proceeds from share issuance   ${cff_shares:>10,.2f}")
print(f"  Proceeds from bank loan        ${cff_loan:>10,.2f}")
print(f"  Dividends paid                 $({-cff_dividend:>9,.2f})")
print(f"  Net Cash from Financing        ${cff_total:>10,.2f}")
print()
print(f"  Net Change in Cash             ${net_change_cash:>10,.2f}")
print(f"  Beginning Cash Balance         ${0:>10,.2f}")
print(f"  Ending Cash Balance            ${net_change_cash:>10,.2f}")
print("=" * 55)
print(f"\nVerification: cash per balance sheet = ${cash:,.2f}")
print(f"Match: {np.isclose(net_change_cash, cash)}")
```

All three statements are consistent and tightly linked:
* Net income of \$25,125 flows into retained earnings on the balance sheet.
* The cash flow statement explains how cash moved from \$0 to \$97,500, reconciling with the balance sheet's cash figure.
* Total assets equal total liabilities plus equity — the accounting equation holds.

> **Key Concept:** This exercise demonstrates the *articulation* of financial statements — they are not independent reports but a tightly linked system. A change in any one statement necessarily ripples through the others.

### How the statements connect

The linkages between statements follow a precise logical chain:

1. **Income statement → Balance sheet:** Net income increases retained earnings (a component of equity). If the company pays dividends, retained earnings = net income − dividends.

2. **Balance sheet → Cash flow statement:** Changes in balance sheet accounts between two periods determine the adjustments in the indirect method of the cash flow statement. For example, if accounts receivable *increased* by \$30,000, that means \$30,000 of revenue was recognised but not yet collected — a cash adjustment.

3. **Cash flow statement → Balance sheet:** The net change in cash from the cash flow statement must equal the change in the cash line on the balance sheet. This is the ultimate check that the three statements are internally consistent.

> **CFA Exam Tip:** If you are given two of the three statements and asked to derive values from the third, use these linkages. For example, if you know net income (from the income statement) and the change in cash (from the cash flow statement), you can deduce the working capital adjustments that bridge the two.

## 8. Visualising Cross-Statement Linkages

The connections between the three statements are easier to grasp visually. Let us create two charts:
1. A **stacked bar chart** showing the accounting equation holds across multiple periods.
2. A **waterfall chart** showing how net income flows from the income statement through retained earnings to the balance sheet.

## Understanding the visualisations

The two charts below serve complementary purposes:

1. **The accounting equation chart** shows the balance sheet's structure evolving over time. It answers: *How is the company financed — by debt (liabilities) or by ownership (equity)?*

2. **The waterfall chart** traces the income statement's flow from revenue to net income, then extends to show how net income flows into retained earnings after dividends. It answers: *Where does the value created by the company end up?*

Together, these visualisations make the cross-statement linkages tangible rather than abstract.

```python
# ── 8a: Accounting equation across 4 quarters ──
# Simulate quarterly snapshots (simplified: assume linear growth)

quarters = ['Q1', 'Q2', 'Q3', 'Q4']
assets      = np.array([150_000, 175_000, 210_000, total_assets])
liabilities = np.array([ 55_000,  58_000,  62_000, total_liabilities])
equity      = assets - liabilities   # A = L + E must hold

x = np.arange(len(quarters))
width = 0.5

fig, ax = plt.subplots(figsize=(10, 6))
bars1 = ax.bar(x, liabilities, width, label='Liabilities', color=SECONDARY, edgecolor='white')
bars2 = ax.bar(x, equity, width, bottom=liabilities, label='Equity', color=PRIMARY, edgecolor='white')

# Mark total assets
for i in range(len(quarters)):
    ax.text(i, assets[i] + 2000, f'${assets[i]:,.0f}', ha='center', fontsize=10, fontweight='bold')

ax.set_xticks(x)
ax.set_xticklabels(quarters)
ax.set_ylabel('Amount ($)')
ax.set_title('Accounting Equation: Assets = Liabilities + Equity')
ax.legend()
plt.tight_layout()
plt.show()
```

In every quarter, the stacked liabilities and equity bars sum exactly to total assets — a visual confirmation that A = L + E holds throughout the year.

Notice the changing composition over time: equity (blue) is growing as the company retains earnings each quarter, while the liability base remains relatively stable. This is a healthy pattern for a young company — it is building equity rather than accumulating debt.

In contrast, a company where the liability stack is growing faster than equity would be increasing its financial leverage — taking on more debt relative to its equity base. Whether this is good or bad depends on the cost of debt, the return on invested capital, and the company's risk tolerance.

> **Key Concept:** The accounting equation chart is a powerful diagnostic tool. By tracking the composition of assets, liabilities, and equity over time, analysts can quickly identify shifts in capital structure, working capital trends, and the balance between internal financing (retained earnings) and external financing (debt or equity issuance).

```python
# ── 8b: Waterfall chart — revenue to net income to retained earnings ──

categories = ['Revenue', 'COGS', 'OpEx', 'Deprec.', 'Interest', 'Tax', 'Net Income', 'Dividends', 'Retained\nEarnings']
values     = [revenue, -cogs, -opex, -depreciation, -interest_exp, -tax_expense, net_income, -dividends, retained_earn]

# Compute waterfall positions
cumulative = np.zeros(len(values))
cumulative[0] = values[0]
for i in range(1, 6):
    cumulative[i] = cumulative[i-1] + values[i]
# Net Income is a subtotal (starts from 0)
cumulative[6] = net_income
# Dividends reduce to retained earnings
cumulative[7] = net_income + values[7]
cumulative[8] = retained_earn

# Bottom of each bar
bottoms = np.zeros(len(values))
for i in range(1, 6):
    bottoms[i] = cumulative[i-1] + min(0, values[i])
bottoms[7] = cumulative[6] + min(0, values[7])

# Bar heights (absolute values)
heights = np.abs(values)
# Subtotals (Net Income, Retained Earnings) start from 0
bottoms[6] = 0
heights[6] = net_income
bottoms[8] = 0
heights[8] = retained_earn

colors = [PRIMARY if v >= 0 else SECONDARY for v in values]
colors[6] = TERTIARY   # subtotal
colors[8] = TERTIARY   # subtotal

fig, ax = plt.subplots(figsize=(12, 6))
bars = ax.bar(categories, heights, bottom=bottoms, color=colors, edgecolor='white', width=0.6)

for bar, val in zip(bars, values):
    y = bar.get_y() + bar.get_height() + 500
    ax.text(bar.get_x() + bar.get_width()/2, y, f'${val:+,.0f}', ha='center', fontsize=9, fontweight='bold')

ax.set_ylabel('Amount ($)')
ax.set_title('Waterfall: Revenue → Net Income → Retained Earnings')
ax.axhline(0, color='grey', linewidth=0.5)
plt.tight_layout()
plt.show()
```

The waterfall chart shows the complete journey from top-line revenue to retained earnings — the amount that ultimately accrues to the balance sheet's equity section. Each red bar represents a deduction (expense or dividend), and the green bars are subtotals.

> **CFA Exam Tip:** Understanding this flow is essential for the "articulation of financial statements" topic. The examiner may give you partial information from one statement and ask you to derive values on another.

### What the waterfall reveals about ExampleCo

The waterfall tells us several things about ExampleCo's economics:

* **COGS is small relative to revenue** — only \$18,000 out of \$80,000, implying a 77.5% gross margin. In a real manufacturing company, this would be unusually high; the numbers are simplified for illustration.
* **Operating expenses (\$20,000) and depreciation (\$6,000)** together represent the overhead of running the business. Together they reduce operating margin to 45%.
* **Interest expense (\$2,500)** is modest — ExampleCo has a conservative capital structure with only \$50,000 of debt.
* **Tax (\$8,375)** at 25% is applied to earnings before tax. The effective tax rate equals the statutory rate because there are no deferred tax differences in this simplified example.
* **Dividends (\$5,000)** are paid from net income, reducing the amount retained in the business. The retention ratio is (\$25,125 - \$5,000) / \$25,125 = 80%.

> **Key Concept:** The *retention ratio* (retained earnings / net income) indicates how much of its profits a company reinvests versus distributes. A high retention ratio funds growth internally; a low retention ratio returns cash to shareholders. Neither is inherently better — it depends on whether the company has profitable reinvestment opportunities.

## 9. IFRS vs US GAAP — Key Differences

The two dominant accounting frameworks are:

* **IFRS** (International Financial Reporting Standards) — used in 140+ countries, issued by the IASB.
* **US GAAP** (Generally Accepted Accounting Principles) — used in the United States, issued by FASB.

While converging in many areas, important differences remain:

| Topic | IFRS | US GAAP |
|-------|------|---------|
| **Framework philosophy** | *Principles-based* — provides broad principles, requires judgment | *Rules-based* — provides specific, detailed rules |
| **Inventory methods** | LIFO is **prohibited** | LIFO is **permitted** |
| **Development costs** | Capitalise if 6 criteria met (IAS 38) | Expense as incurred (except software: ASC 350-40) |
| **Revaluation of assets** | Permitted (upward revaluation of PP&E, intangibles) | Not permitted (historical cost only) |
| **Impairment reversal** | Permitted for most assets (not goodwill) | Not permitted |
| **Revenue standard** | IFRS 15 | ASC 606 (largely converged) |
| **Lease classification** | Single model — all leases on balance sheet (IFRS 16) | Two models — finance vs operating (ASC 842) |
| **Income statement format** | Either function or nature of expense | Function of expense only |
| **Cash flow — interest paid** | Operating or financing (choice) | Operating only |
| **Cash flow — dividends paid** | Operating or financing (choice) | Financing only |

> **Key Concept:** The IFRS/GAAP distinction matters for cross-border comparisons. A European company using IFRS and a US company using GAAP may report different earnings for identical economic transactions — not because they did anything different, but because the *accounting rules* differ.

> **CFA Exam Tip:** The CFA exam tests specific differences — particularly inventory (LIFO), development costs, asset revaluation, impairment reversal, and cash flow classifications. The table above covers the highest-frequency exam topics.

### Practical implications for analysts

When comparing a European company (IFRS) with a US company (GAAP), the following adjustments are commonly needed:

1. **Inventory:** If the US company uses LIFO, add back the LIFO reserve to inventory and equity (after tax) to make it comparable to the IFRS company's FIFO inventory.

2. **Development costs:** If the IFRS company capitalises development costs, these appear as intangible assets on the balance sheet rather than as R&D expense on the income statement. To compare on an equal basis, either capitalise the US company's R&D or expense the IFRS company's capitalised development.

3. **Asset revaluation:** If the IFRS company has revalued PP&E upward, its assets and equity will be higher, and its depreciation expense will be higher (depreciating from a higher base). Analysts may need to reverse the revaluation to compare on a historical-cost basis.

4. **Cash flow classification:** Interest paid classified as financing (permitted under IFRS) rather than operating (required under GAAP) makes IFRS CFO look *higher*. Reclassify for comparability.

> **Key Concept:** The differences are not just theoretical — they create measurable distortions in financial ratios. A study by Barth et al. (2012) found that converting IFRS firms to US GAAP equivalents changed ROE by an average of 2-3 percentage points. For analysts managing global portfolios, these adjustments are not optional.

## 10. The Auditor's Report

### Purpose and scope

An independent external auditor examines the company's financial statements and issues an **opinion** on whether they present a *true and fair view* (IFRS) or are *free from material misstatement* (US GAAP).

### Types of audit opinions

| Opinion | Meaning | Investor Implication |
|---------|---------|---------------------|
| **Unqualified (clean)** | Statements are fairly presented in all material respects | Highest assurance — no red flags |
| **Qualified** | Fairly presented *except for* a specific issue | Investigate the exception |
| **Adverse** | Statements are *not* fairly presented | Serious concern — do not rely on the statements |
| **Disclaimer** | Auditor *cannot* form an opinion (scope limitation) | Extreme caution |

> **Common Mistake:** An unqualified ("clean") opinion does **not** guarantee the company is healthy or profitable. It only means the *accounting* is materially correct. A company can have a clean audit and still be on the brink of bankruptcy.

### Going concern

If the auditor has doubts about the company's ability to continue as a **going concern** (i.e. survive for the next 12 months), they must disclose this in an *emphasis of matter* paragraph. This is one of the most important signals an analyst can receive.

### Limitations of auditing

* Audits test a *sample* of transactions, not every single one.
* Audits provide **reasonable assurance**, not absolute assurance.
* Fraud, especially management-level fraud with collusion, may evade audit procedures.
* The expectation gap: users expect audits to catch all fraud; auditors are designed to catch *material misstatements*.

### The audit process in brief

The audit is not a line-by-line check of every transaction. Instead, auditors use a **risk-based approach**:

1. **Planning:** Identify areas of highest misstatement risk (complex estimates, related-party transactions, revenue recognition).
2. **Testing:** Design procedures to gather evidence — sampling transactions, confirming balances with third parties, analytical procedures.
3. **Evaluation:** Assess whether identified misstatements, individually or in aggregate, are **material** (large enough to influence the decisions of a reasonable investor).
4. **Reporting:** Issue the audit opinion based on the evidence gathered.

> **CFA Exam Tip:** Materiality is a quantitative *and* qualitative concept. A \$1 million misstatement might be immaterial for a \$50 billion company but could be material if it turns a reported profit into a loss, triggers a debt covenant violation, or involves fraud by senior management.

### Historical context: landmark audit failures

Understanding the auditor's limitations is easier with concrete examples:

* **Enron (2001):** Arthur Andersen issued clean opinions for years while Enron used off-balance-sheet entities to hide billions in debt. This led to the Sarbanes-Oxley Act (SOX) in 2002, which created the PCAOB to oversee auditors and required management to certify the accuracy of financial statements.

* **WorldCom (2002):** The company capitalised billions of dollars in ordinary operating expenses (line costs), inflating assets and understating expenses. The fraud was eventually discovered by an internal auditor, not the external audit firm.

* **Wirecard (2020):** The German fintech company fabricated \$2.1 billion in cash balances. The auditor (EY) was unable to obtain third-party confirmations for the cash but continued to issue clean opinions for years.

These cases illustrate a consistent pattern: the most damaging frauds involve **collusion at the senior management level**, which is precisely the scenario that standard audit procedures are least equipped to detect.

> **Key Concept:** The expectation gap — the difference between what the public *expects* auditors to do (catch all fraud) and what auditors are actually designed to do (provide reasonable assurance against material misstatement) — has persisted for decades despite regulatory reforms. Sophisticated analysts treat the audit opinion as a necessary but not sufficient condition for trusting financial statements.

## 11. Limitations of Financial Statements

No introduction to financial statements is complete without understanding their inherent limitations:

### 11.1 Historical cost bias

Most items on the balance sheet are recorded at *historical cost* (what the company originally paid), not current market value. A building purchased in 1990 for \$1 million may be worth \$10 million today, but under US GAAP the balance sheet still shows \$1 million (less depreciation).

> **Key Concept:** IFRS allows revaluation to fair value for some assets, partially addressing this limitation. US GAAP does not — another reason cross-border comparisons require adjustments.

### 11.2 Estimates and judgment

Financial statements are full of estimates:
* **Allowance for doubtful accounts** — how much of accounts receivable will never be collected?
* **Useful life of assets** — is a machine's life 5 years or 15 years?
* **Warranty provisions** — how many products will be returned?
* **Pension obligations** — what discount rate to use?

These estimates can be honest differences of judgment or, at worst, vehicles for earnings manipulation.

### 11.3 Off-balance-sheet items

Certain obligations may not appear on the balance sheet:
* **Operating leases** (largely resolved by IFRS 16 / ASC 842, but legacy statements still exist).
* **Special purpose entities** (SPEs / VIEs) — used infamously by Enron.
* **Contingent liabilities** — lawsuits, environmental cleanup — disclosed in footnotes but not recognised unless "probable".

### 11.4 Intangible value

Internally developed brands, customer relationships, workforce quality, and corporate culture — often a company's most valuable assets — do **not** appear on the balance sheet (only *purchased* intangibles are recognised).

> **CFA Exam Tip:** The limitations above motivate the *adjustments* that analysts make to reported numbers — a recurring theme throughout the CFA curriculum. Understanding what financial statements *cannot* tell you is just as important as understanding what they *can*.

### 11.5 Comparability challenges

Even within the same accounting framework, companies make different choices about:
* Depreciation methods and useful lives
* Inventory cost flow assumptions
* Revenue recognition timing (especially for multi-element arrangements)
* Capitalisation vs expensing thresholds

These choices are all "within GAAP" (or IFRS) but can make two otherwise identical companies look very different on paper. The analyst's job is to **normalise** — adjust reported numbers to a common basis for fair comparison.

> **CFA Exam Tip:** The CFA curriculum identifies four desirable qualities of financial reports: **relevance**, **faithful representation**, **comparability**, and **timeliness**. Limitations 11.1–11.5 above show how real-world financial statements fall short of these ideals — and why financial analysis (rather than just financial *reading*) is necessary.
