### Why This Topic Matters

If someone offered you **$1,000 today** or **$1,000 one year from now**, which would you choose? Almost everyone would take the money today. But *why*? There are three fundamental reasons:

1. **Opportunity cost:** If you have $1,000 today, you can invest it and earn a return. That $1,000 could become $1,080 in a year at 8% interest. The money tomorrow has *no* earning potential between now and then.
2. **Inflation:** Prices tend to rise over time. $1,000 today buys more goods and services than $1,000 will buy next year.
3. **Risk/uncertainty:** A promise of future money carries the risk that it might not materialize. The bird in hand is worth more than one in the bush.

This simple insight -- that **a dollar today is worth more than a dollar tomorrow** -- is the foundation of *all* of finance. Every bond price, stock valuation, loan payment, and corporate investment decision rests on this principle.

> **Key Concept:** The **Time Value of Money (TVM)** states that money available now is worth more than the same amount in the future, due to its potential earning capacity. This is the most fundamental concept in finance.

Think of it this way: money is like a seed. If you plant it today (invest it), it grows over time. A seed sitting in a warehouse for a year doesn't grow at all. That's the time value of money -- money's ability to grow when given time.

## 1. Simple vs Compound Interest

When you lend money (or deposit it in a bank), you earn interest as compensation for giving up the use of your money. There are two fundamentally different ways interest can be calculated.

### Simple Interest: The Straightforward Approach

With **simple interest**, you earn interest *only* on the original amount you invested (the **principal**). The interest earned each period is the same fixed amount, no matter how long the investment lasts.

**The analogy:** Think of simple interest like a paycheck. You work (lend your money), and you get paid a fixed amount each period. Last year's paychecks don't earn you more paychecks.

**The formula:**

$$FV_{\text{simple}} = PV \cdot (1 + r \cdot n)$$

where:
- $FV$ = Future Value (what your money grows to)
- $PV$ = Present Value (your original investment / principal)
- $r$ = interest rate per period (as a decimal, so 8% = 0.08)
- $n$ = number of periods

**Worked Example:** You invest **$1,000** at **8% simple interest** for **3 years**.
- Year 1 interest: $1,000 x 0.08 = $80
- Year 2 interest: $1,000 x 0.08 = $80 (still on original $1,000)
- Year 3 interest: $1,000 x 0.08 = $80
- Total: $1,000 + $240 = **$1,240**
- Using the formula: $1,000 x (1 + 0.08 x 3) = $1,000 x 1.24 = **$1,240** -- matches!

### Compound Interest: Interest on Interest

With **compound interest**, you earn interest on both the principal *and* on any interest previously earned. This is the mechanism behind most real-world investments.

**The analogy:** Compound interest is like a snowball rolling downhill. As it rolls, it picks up more snow. The bigger it gets, the more snow it picks up with each revolution. The snowball doesn't just grow at a constant rate -- it *accelerates*. That's the magic of compounding.

**The formula:**

$$FV_{\text{compound}} = PV \cdot (1 + r)^n$$

where the variables mean the same thing, but now $(1+r)$ is raised to the power $n$ instead of multiplied by $n$.

**Worked Example:** You invest **$1,000** at **8% compound interest** for **3 years**.
- Year 1: $1,000 x 1.08 = **$1,080.00** (interest = $80.00)
- Year 2: $1,080 x 1.08 = **$1,166.40** (interest = $86.40 -- more than year 1!)
- Year 3: $1,166.40 x 1.08 = **$1,259.71** (interest = $93.31)
- Using the formula: $1,000 x (1.08)^3 = $1,000 x 1.259712 = **$1,259.71**

The compound interest investor earned **$19.71 more** than the simple interest investor ($1,259.71 vs $1,240.00). That extra amount is the "interest on interest" -- the interest earned in year 2 on the $80 from year 1, and so on.

### The Power of Compounding Over Time

The difference between simple and compound interest may seem small over 3 years, but it becomes enormous over longer periods. This is the key insight that Albert Einstein (apocryphally) called "the eighth wonder of the world."

> **Key Concept:** The difference between simple and compound interest grows **exponentially** over time. Over short periods, the difference is negligible. Over decades, it becomes the dominant factor in wealth accumulation.

> **CFA Exam Tip:** Unless stated otherwise, always assume **compound interest** on the CFA exam. Simple interest is used mainly for short-term instruments like Treasury bills and money market instruments.

Let's verify our hand calculations and see the dramatic divergence over 50 years.

```python
def simple_interest_fv(pv, r, n):
    """Future value with simple interest: FV = PV * (1 + r*n)"""
    return pv * (1 + r * n)


def compound_interest_fv(pv, r, n):
    """Future value with compound interest: FV = PV * (1 + r)^n"""
    return pv * (1 + r) ** n


# ── Worked example: $1,000 at 8% for 3 years
pv, r, n = 1_000, 0.08, 3
fv_simple = simple_interest_fv(pv, r, n)
fv_compound = compound_interest_fv(pv, r, n)

print(f"Initial investment: ${pv:,.0f}")
print(f"Rate: {r:.0%}, Period: {n} years\n")
print(f"Simple interest FV:   ${fv_simple:>10,.2f}")
print(f"Compound interest FV: ${fv_compound:>10,.2f}")
print(f"Difference:           ${fv_compound - fv_simple:>10,.2f} (interest on interest)")
```

The code confirms our hand calculations: $1,259.71 for compound vs $1,240.00 for simple. The $19.71 difference is small over 3 years, but watch what happens as we extend the time horizon.

The difference is small over 5 years but compounds dramatically over longer horizons.

## 2. Compounding Frequency and Effective Annual Rate (EAR)

### The Problem with Stated Rates

Banks and financial institutions often quote interest rates with different compounding frequencies. A savings account might advertise "6% compounded monthly" while a bond pays "6% compounded semi-annually." Are these the same? **No!** And this is a common source of confusion.

The **stated rate** (also called the nominal rate or Annual Percentage Rate, APR) is the rate *before* accounting for compounding within the year. To make fair comparisons, we need to convert everything to the same basis: the **Effective Annual Rate (EAR)**.

### The Intuition

Why does more frequent compounding give you more money? Because each time interest is compounded, it starts earning interest on the interest from the previous compounding period. Monthly compounding means you start earning "interest on interest" after just one month, while annual compounding makes you wait a full year.

Think of it this way: if you compound 12 times per year instead of once, you get 12 chances to start the snowball rolling faster.

### The Formula

$$\text{EAR} = \left(1 + \frac{r_s}{m}\right)^m - 1$$

where:
- $r_s$ = stated (nominal) annual rate
- $m$ = number of compounding periods per year

**Worked Example:** A bank advertises **6% compounded monthly**. What is the EAR?

$$\text{EAR} = \left(1 + \frac{0.06}{12}\right)^{12} - 1 = (1.005)^{12} - 1 = 1.06168 - 1 = 6.168\%$$

So 6% compounded monthly is equivalent to **6.168%** compounded annually. The extra 0.168% is the boost from within-year compounding.

### Continuous Compounding: The Theoretical Limit

What if we compound every second? Every nanosecond? As $m \to \infty$:

$$\text{EAR}_{\text{continuous}} = e^{r_s} - 1$$

This uses the fundamental limit from calculus: $\lim_{m \to \infty} (1 + r_s/m)^m = e^{r_s}$.

Continuous compounding is the theoretical maximum -- you can't earn more by compounding more frequently. In practice, daily compounding gets you 99.9% of the way to continuous.

**Worked Example:** What is the EAR for **6% compounded continuously**?

$$\text{EAR} = e^{0.06} - 1 = 1.06184 - 1 = 6.184\%$$

Compare this to monthly compounding (6.168%) -- the difference is only 0.016%. Continuous compounding is a mathematical convenience more than a practical difference.

### Comparison Table: Same 6% Stated Rate, Different Compounding

| Compounding | $m$ | EAR | Extra vs Annual |
|---|---|---|---|
| Annual | 1 | 6.000% | -- |
| Semi-annual | 2 | 6.090% | +0.090% |
| Quarterly | 4 | 6.136% | +0.136% |
| Monthly | 12 | 6.168% | +0.168% |
| Daily | 365 | 6.183% | +0.183% |
| Continuous | -- | 6.184% | +0.184% |

Notice the **diminishing returns**: the jump from annual to semi-annual is larger than the jump from monthly to daily.

> **CFA Exam Tip:** When comparing investments with different compounding frequencies, always convert to EAR first. The investment with the higher EAR is the better deal, regardless of the stated rate.

> **Common Mistake:** Don't confuse the stated rate with the periodic rate. If a bond pays 6% semi-annually, the periodic rate is 3% (not 6%), and you compound twice per year.

Let's verify these calculations and visualize how EAR changes with compounding frequency.

```python
def ear(stated_rate, m):
    """Effective Annual Rate: EAR = (1 + r_s/m)^m - 1"""
    return (1 + stated_rate / m) ** m - 1


def ear_continuous(stated_rate):
    """Continuous compounding: EAR = e^r - 1"""
    return np.exp(stated_rate) - 1


# ── Compare compounding frequencies for 6% stated rate
r_s = 0.06
frequencies = [
    ('Annual', 1), ('Semi-annual', 2), ('Quarterly', 4),
    ('Monthly', 12), ('Daily', 365),
]

print(f"Stated rate: {r_s:.0%}\n")
print(f"{'Compounding':<15} {'m':>5} {'EAR':>10} {'Extra vs Annual':>16}")
print("-" * 48)
ear_annual = ear(r_s, 1)
for name, m in frequencies:
    e = ear(r_s, m)
    print(f"{name:<15} {m:>5} {e:>10.4%} {(e - ear_annual):>15.4%}")

e_cont = ear_continuous(r_s)
print(f"{'Continuous':<15} {'inf':>5} {e_cont:>10.4%} {(e_cont - ear_annual):>15.4%}")
```

The output confirms our hand calculations. Notice the diminishing returns:
- Annual to semi-annual: a jump of about 0.090%
- Semi-annual to quarterly: a smaller jump of about 0.046%
- Daily to continuous: a negligible difference of about 0.001%

This confirms our intuition: continuous compounding is the theoretical ceiling, and daily compounding already gets you 99.9% of the way there.

The following chart makes the diminishing returns visually clear.The output confirms our hand calculations. Notice the diminishing returns:

* Annual → Semi-annual: meaningful jump in EAR
* Semi-annual → Quarterly: smaller increase
* Monthly → Daily: barely any difference
* Daily → Continuous: virtually no additional benefit

### The mathematical limit — continuous compounding

As compounding frequency $m \to \infty$, the EAR approaches a finite limit:

$$\lim_{m \to \infty} \left(1 + \frac{r}{m}\right)^m = e^r$$

This is why **continuous compounding** uses the formula:

$$\text{EAR} = e^r - 1$$

The number $e \approx 2.71828$ arises naturally as this limit. The convergence is rapid: monthly compounding gets you most of the way there, daily compounding gets you essentially all the way.

| Compounding | EAR (at 6% stated) |
|-------------|--------------------|
| Annual | 6.0000% |
| Semi-annual | 6.0900% |
| Quarterly | 6.1364% |
| Monthly | 6.1678% |
| Daily | 6.1831% |
| Continuous | 6.1837% |

> **CFA Exam Tip:** The EAR formula is a frequently tested concept. Remember: $\text{EAR} = (1 + r/m)^m - 1$. The EAR allows you to compare investments with different compounding frequencies on an apples-to-apples basis. Always convert to EAR before comparing rates.

```python
# ── Visualization: EAR vs compounding frequency
m_values = np.arange(1, 366)
ear_values = [ear(r_s, m) for m in m_values]

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(m_values, np.array(ear_values) * 100, color=PRIMARY, linewidth=2)
ax.axhline(ear_continuous(r_s) * 100, color=SECONDARY, linestyle='--', linewidth=2,
           label=f'Continuous: {ear_continuous(r_s):.4%}')

# Mark key frequencies
for name, m in frequencies:
    e = ear(r_s, m)
    ax.plot(m, e * 100, 'o', color=TERTIARY, markersize=8, zorder=5)
    ax.annotate(f'{name}\n{e:.3%}', (m, e * 100), textcoords='offset points',
               xytext=(15, -10), fontsize=9)

ax.set_xlabel('Compounding Periods per Year (m)')
ax.set_ylabel('Effective Annual Rate (%)')
ax.set_title(f'EAR vs Compounding Frequency (Stated Rate = {r_s:.0%})')
ax.legend()
plt.tight_layout()
plt.show()
```

This has a practical implication: when comparing a monthly-compounded rate to a daily-compounded rate, the difference is negligible. But comparing annual to semi-annual compounding can make a meaningful difference, especially on large loan amounts.**Reading the chart:** The EAR rises steeply from annual to monthly compounding, then flattens dramatically. ## 3. Present Value & Future Value

### The Two Sides of the Same Coin

Present Value and Future Value are really just two ways of looking at the same relationship. If compound interest lets us move money **forward in time** (PV to FV), then discounting lets us move money **backward in time** (FV to PV).

$$FV = PV \cdot (1 + r)^n \qquad \Longleftrightarrow \qquad PV = \frac{FV}{(1 + r)^n}$$

where:
- $FV$ = Future Value (what money grows to)
- $PV$ = Present Value (what future money is worth today)
- $r$ = discount rate (or required rate of return)
- $n$ = number of periods

### The Discount Factor

The term $\frac{1}{(1+r)^n}$ is called the **discount factor**. It tells you how many cents on the dollar a future payment is worth today.

**Worked Example:** What is $10,000 received in **10 years** worth today if your required return is **8%**?

$$PV = \frac{\$10{,}000}{(1.08)^{10}} = \frac{\$10{,}000}{2.15892} = \$4{,}631.93$$

The discount factor is $\frac{1}{2.15892} = 0.4632$. In other words, at an 8% discount rate, a dollar received in 10 years is worth only about **46 cents** today.

**Why does this matter for investors?** When you buy a bond, you're paying today's price for a series of future cash flows. The present value calculation tells you the *maximum* price you should pay. If the bond's price is less than the PV of its cash flows, it's a good deal.

### What Affects Present Value?

Two factors reduce present value:
1. **Higher discount rates** -- the more you could earn elsewhere, the less future money is worth to you today
2. **Longer time horizons** -- money further in the future has more time to "shrink" through discounting

> **Key Concept:** Discounting is the reverse of compounding. Compounding grows money forward in time; discounting shrinks money backward in time. They are mathematical inverses.

> **CFA Exam Tip:** Remember the relationship: PV and discount rate move in **opposite** directions (higher rate = lower PV). PV and time also move in opposite directions (more time = lower PV). This is why long-duration bonds are more sensitive to interest rate changes.

Let's calculate PV at different rates and visualize how these factors interact.

```python
def future_value(pv, r, n):
    """Compute future value: FV = PV * (1+r)^n"""
    return pv * (1 + r) ** n


def present_value(fv, r, n):
    """Compute present value: PV = FV / (1+r)^n"""
    return fv / (1 + r) ** n


def discount_factor(r, n):
    """Compute the discount factor 1/(1+r)^n."""
    return 1 / (1 + r) ** n


# ── Example: PV of $10,000 received in 10 years at different rates
fv_target = 10_000
rates = [0.04, 0.06, 0.08, 0.10]
n_years = 10

print(f"Present Value of ${fv_target:,.0f} received in {n_years} years:\n")
for r in rates:
    pv = present_value(fv_target, r, n_years)
    df = discount_factor(r, n_years)
    print(f"  r = {r:.0%}: PV = ${pv:>8,.2f}  (discount factor = {df:.4f})")
```

**Interpreting the output:** Look at how much the present value drops as the discount rate increases:
- At 4%: $10,000 in 10 years is worth $6,756 today
- At 10%: the same $10,000 is worth only $3,855 today

That's nearly a **$3,000 difference** just from changing the discount rate. This is why interest rates are so important in finance -- when rates rise, the present value of all future cash flows falls, which is why bond prices drop when interest rates increase.

The following charts show this relationship visually.**Interpreting the output:** Look at how much the present value drops as the discount rate rises. At a 5% discount rate, \$1,000 in 10 years is worth \$613.91 today. At 15%, the same future amount is worth only \$247.18 — less than half.

### The discount rate is the most important assumption

In any present value calculation, the discount rate has the largest single impact on the result. Small changes in the discount rate produce large changes in PV:

| Discount Rate | PV of \$1,000 in 10 years | Change |
|---------------|---------------------------|--------|
| 4% | \$675.56 | baseline |
| 6% | \$558.39 | −17.4% |
| 8% | \$463.19 | −31.4% |
| 10% | \$385.54 | −42.9% |
| 12% | \$321.97 | −52.3% |

### What determines the right discount rate?

The discount rate should reflect the **risk-adjusted opportunity cost** of capital:

* **Risk-free rate** (e.g., government bond yield) for guaranteed cash flows
* **Cost of debt** for fixed-income investments (corporate bonds)
* **Cost of equity** (CAPM-derived) for equity investments
* **WACC** (weighted average cost of capital) for whole-firm enterprise valuation

Higher risk demands a higher discount rate, which lowers PV. This is why cash flows from a stable utility are valued more highly per dollar than identical cash flows from a volatile startup.

> **CFA Exam Tip:** The CFA exam frequently tests sensitivity to discount rate. If asked to evaluate a project's NPV under uncertainty, always compute NPV at multiple discount rates. The "break-even discount rate" — where NPV = 0 — is the project's IRR (Internal Rate of Return), the maximum discount rate at which the project still makes economic sense.

```python
# ── Visualization: PV as a function of time for different rates
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: PV vs years for different rates
years_range = np.arange(0, 31)
colors = [PRIMARY, SECONDARY, TERTIARY, ACCENT]
for r_val, color in zip(rates, colors):
    pvs = [present_value(fv_target, r_val, n) for n in years_range]
    axes[0].plot(years_range, pvs, color=color, linewidth=2, label=f'r = {r_val:.0%}')
axes[0].set_xlabel('Years Until Payment')
axes[0].set_ylabel('Present Value ($)')
axes[0].set_title(f'PV of ${fv_target:,.0f} at Different Discount Rates')
axes[0].legend()

# Right: FV growth from $1,000
pv_start = 1_000
for r_val, color in zip(rates, colors):
    fvs = [future_value(pv_start, r_val, n) for n in years_range]
    axes[1].plot(years_range, fvs, color=color, linewidth=2, label=f'r = {r_val:.0%}')
axes[1].set_xlabel('Years')
axes[1].set_ylabel('Future Value ($)')
axes[1].set_title(f'Growth of ${pv_start:,.0f} at Different Rates')
axes[1].legend()

plt.tight_layout()
plt.show()
```

**Reading the charts:**

- **Left chart (PV):** All curves start at $10,000 (when payment is immediate, PV = FV). As the payment moves further into the future, the PV shrinks. Higher rates cause steeper declines -- at 10%, money 30 years away is worth almost nothing today.

- **Right chart (FV):** This is the mirror image. Starting from $1,000, higher rates produce steeper growth curves. The 10% curve dramatically outpaces the 4% curve over 30 years.

These two charts are really showing the same mathematical relationship from opposite perspectives. They illustrate a profound truth: the discount rate is the bridge between the present and the future in all of finance.**Reading the charts:**

The first chart shows PV as a function of time for various discount rates. Three patterns emerge:

1. **All curves slope downward:** PV always decreases as the future date moves further out. The longer you wait, the less today's claim is worth.

2. **Higher discount rates produce steeper curves:** A 12% rate discounts future cash flows much more aggressively than a 4% rate. This is why high-rate environments depress asset prices broadly.

3. **The curves are convex (curve upward):** Each additional year of delay reduces PV less than the previous year. Mathematically, this is because $(1+r)^n$ grows exponentially in $n$, so $1/(1+r)^n$ decays toward zero more slowly with each step.

### The "rule of 72" for present value

The rule of 72 estimates the doubling/halving time:

$$\text{Time to halve PV} \approx \frac{72}{r \times 100}$$

At 6% discount rate, PV halves every ~12 years. At 9%, every ~8 years. At 12%, every ~6 years.

This rough heuristic helps with quick mental math: "If I delay receiving \$10,000 by 12 years at a 6% discount rate, it's worth about \$5,000 today." Useful for cocktail-napkin financial reasoning.

> **Common Mistake:** Students sometimes think PV calculations require complex spreadsheets. They don't. The rule of 72 (or its variant for tripling: rule of 110) gives 5%-accurate estimates instantly. Mastering these heuristics is more valuable than having a calculator handy.

## 4. Annuities

### What Is an Annuity?

An **annuity** is a series of **equal cash flows** occurring at **regular intervals** for a fixed number of periods. Annuities are everywhere in finance:

- **Mortgage payments** -- equal monthly payments for 15 or 30 years
- **Car loan payments** -- equal monthly payments for 3-7 years
- **Retirement withdrawals** -- equal annual withdrawals from a retirement account
- **Bond coupon payments** -- equal semi-annual interest payments

There are two types:

| Type | Payments Occur | Example |
|---|---|---|
| **Ordinary annuity** | End of each period | Mortgage payments, bond coupons |
| **Annuity due** | Beginning of each period | Rent payments, insurance premiums |

### Ordinary Annuity: Present Value

Suppose you receive $C$ at the end of each period for $n$ periods. What is the total present value?

**The brute force approach:** Discount each payment individually and add them up:

$$PV = \frac{C}{(1+r)^1} + \frac{C}{(1+r)^2} + \cdots + \frac{C}{(1+r)^n} = C \sum_{t=1}^{n} \frac{1}{(1+r)^t}$$

This works, but it's tedious for large $n$. Fortunately, this is a **geometric series**, and there's a closed-form formula.

**The closed-form formula (derived from geometric series):**

$$PV_{\text{ordinary}} = C \cdot \frac{1 - (1+r)^{-n}}{r}$$

The term $\frac{1 - (1+r)^{-n}}{r}$ is called the **present value annuity factor**.

*Proof:* Let $x = \frac{1}{1+r}$. Then $S = \sum_{t=1}^{n} x^t = x \cdot \frac{1 - x^n}{1 - x}$. Substituting back and simplifying yields the formula above.

**Worked Example:** You will receive **$1,000 per year** for **20 years**, and the discount rate is **6%**. What is the present value of this annuity?

$$PV = 1{,}000 \times \frac{1 - (1.06)^{-20}}{0.06} = 1{,}000 \times \frac{1 - 0.3118}{0.06} = 1{,}000 \times 11.4699 = \$11{,}469.92$$

So 20 payments of $1,000 are worth **$11,469.92** today. Notice this is significantly less than the simple sum of $20,000 -- the difference is the time value of money at work.

### Annuity Due: Payments at the Beginning

An annuity due is identical to an ordinary annuity, except each payment arrives **one period sooner**. Since each payment is discounted one less period, each is worth $(1+r)$ times more:

$$PV_{\text{due}} = PV_{\text{ordinary}} \times (1 + r)$$

### Future Value of an Annuity

If you save $C$ per period for $n$ periods, how much will you have at the end?

$$FV_{\text{ordinary}} = C \cdot \frac{(1+r)^n - 1}{r}$$

**Worked Example:** You save **$500 per month** for **30 years** at **7% annual** (monthly rate = 0.583%).

$$FV = 500 \times \frac{(1.00583)^{360} - 1}{0.00583} \approx \$609{,}985$$

You deposited only $500 x 360 = $180,000 total, but compounding turned it into nearly **$610,000**! Over $430,000 came from investment returns.

> **Key Concept:** The annuity PV formula is simply the sum of a geometric series of discount factors. You don't need to memorize it as a separate formula -- it's a direct consequence of the basic PV formula applied to equal cash flows.

> **CFA Exam Tip:** Make sure you know whether a problem describes an ordinary annuity or an annuity due. The telltale: ordinary annuities say payments are at the "end" of the period; annuity due says "beginning." Getting this wrong shifts every answer by a factor of $(1+r)$.

> **Common Mistake:** Forgetting to match the compounding frequency to the payment frequency. If you make monthly payments but the rate is quoted annually, you must divide the annual rate by 12 and multiply the number of years by 12.

Let's verify these formulas with code.

```python
def annuity_pv(C, r, n, due=False):
    """Present value of an annuity.
    
    Parameters
    ----------
    C   : float -- periodic cash flow
    r   : float -- discount rate per period
    n   : int   -- number of periods
    due : bool  -- if True, annuity due (payments at start)
    """
    if r == 0:
        pv = C * n
    else:
        pv = C * (1 - (1 + r) ** (-n)) / r
    return pv * (1 + r) if due else pv


def annuity_fv(C, r, n, due=False):
    """Future value of an annuity."""
    if r == 0:
        fv = C * n
    else:
        fv = C * ((1 + r) ** n - 1) / r
    return fv * (1 + r) if due else fv


# ── Verify: PV via formula vs brute-force summation
C, r, n = 1_000, 0.06, 20

pv_formula = annuity_pv(C, r, n)
pv_brute = sum(C / (1 + r) ** t for t in range(1, n + 1))

print(f"Ordinary annuity: ${C:,.0f}/year for {n} years at {r:.0%}")
print(f"  PV (formula):     ${pv_formula:>10,.2f}")
print(f"  PV (brute-force): ${pv_brute:>10,.2f}")
print(f"  Match: {np.isclose(pv_formula, pv_brute)}")

# ── Annuity due vs ordinary
pv_due = annuity_pv(C, r, n, due=True)
print(f"\n  PV (annuity due): ${pv_due:>10,.2f}")
print(f"  Ratio due/ordinary: {pv_due / pv_formula:.4f} = 1 + r = {1 + r:.4f}")

# ── Future values
fv_ord = annuity_fv(C, r, n)
fv_due = annuity_fv(C, r, n, due=True)
print(f"\n  FV (ordinary):    ${fv_ord:>10,.2f}")
print(f"  FV (annuity due): ${fv_due:>10,.2f}")
```

**What the output tells us:**

1. **Formula vs brute-force match perfectly** -- this confirms our geometric series derivation is correct.
2. **Annuity due is worth exactly (1 + r) times the ordinary annuity** -- since each payment arrives one period sooner, each is worth 6% more, so the whole annuity is worth 6% more.
3. **The FV tells us what a savings plan accumulates to** -- if you save $1,000 per year for 20 years at 6%, you'll have $36,786. That's $16,786 more than the $20,000 you deposited, all from compound interest.**What the output tells us:**

The annuity calculations reveal several important relationships:

1. **PV of an annuity is finite even for many periods:** Despite paying \$100 every year for 30 years (total payments of \$3,000), the PV is only \$1,289.86 at a 5% discount rate. The discounting drastically reduces the value of distant payments.

2. **Annuity due > ordinary annuity:** When payments come at the *beginning* of the period (annuity due), each payment is discounted one fewer period. The PV is higher by exactly a factor of $(1+r)$.

3. **The annuity factor matters:** The PV of an annuity equals the periodic payment times the **annuity factor** $\frac{1-(1+r)^{-n}}{r}$. This factor depends only on the discount rate and number of periods — not on the payment amount. Once you have the factor, scaling to different payment sizes is trivial.

### Practical applications

Annuity formulas appear everywhere in finance:

* **Bond pricing:** Coupon payments are an annuity (the principal at maturity is a separate lump sum).
* **Mortgage payments:** Calculate the level payment that fully amortises a loan.
* **Pensions:** Convert lump-sum savings into a guaranteed monthly income stream.
* **Lottery winnings:** Compare the lump-sum option to the annual annuity option.

> **CFA Exam Tip:** The most commonly tested annuity calculation is finding the periodic payment given PV (or FV), the rate, and the number of periods. This is solving the annuity formula for the payment $C$, given by $C = \frac{PV \cdot r}{1 - (1+r)^{-n}}$. Memorise this formula or, better, derive it on the fly from the basic annuity PV formula.

## 5. Perpetuities

### What If the Payments Never Stop?

A **perpetuity** is an annuity that pays forever -- an infinite stream of cash flows. This might sound like it should have infinite value, but it doesn't! Because of the time value of money, payments far in the future are worth almost nothing today.

**The analogy:** Imagine stacking thinner and thinner pancakes on top of each other. Even though you keep adding pancakes forever, the stack has a finite height because each additional pancake is thinner than the last.

### Constant Perpetuity

$$PV = \frac{C}{r}$$

where:
- $C$ = constant periodic cash flow
- $r$ = discount rate

*Derivation:* Start with the annuity formula and let $n \to \infty$:
$$\lim_{n \to \infty} C \cdot \frac{1 - (1+r)^{-n}}{r} = \frac{C}{r}$$
since $(1+r)^{-n} \to 0$ as $n \to \infty$ (for $r > 0$).

**Worked Example:** A preferred stock pays a **$5 dividend** forever. If your required return is **10%**:

$$PV = \frac{\$5}{0.10} = \$50$$

You should be willing to pay up to **$50** for this stock. This formula is beautifully simple and intuitive: if you invest $50 at 10%, you earn $5 per year forever, exactly replicating the perpetuity.

### Growing Perpetuity (The Gordon Growth Model)

What if cash flows **grow** at a constant rate $g$ per period? If the first cash flow is $C$ and it grows at rate $g < r$:

$$PV = \frac{C}{r - g}$$

This is the famous **Gordon Growth Model** (or Dividend Discount Model), widely used for stock valuation.

**Worked Example:** A stock is expected to pay a **$3.00 dividend** next year, growing at **4% per year** forever. If the required return on equity is **10%**:

$$\text{Stock Price} = \frac{\$3.00}{0.10 - 0.04} = \frac{\$3.00}{0.06} = \$50.00$$

> **Common Mistake:** The growth rate $g$ must be **strictly less than** the discount rate $r$. If $g \geq r$, the formula breaks down (gives infinite or negative values). In practice, no company can grow faster than the economy forever.

> **CFA Exam Tip:** The Gordon Growth Model is appropriate for mature, stable companies with predictable dividend growth (e.g., utilities, consumer staples). It is **not** appropriate for high-growth companies whose growth rate may exceed the required return.

```python
def perpetuity_pv(C, r, g=0):
    """Present value of a (growing) perpetuity.
    
    Parameters
    ----------
    C : float -- first cash flow (at t=1)
    r : float -- discount rate
    g : float -- growth rate (must be < r)
    """
    assert r > g, "Discount rate must exceed growth rate"
    return C / (r - g)


# ── Example: Stock valuation with Gordon Growth Model
dividend = 3.00       # next year's dividend
cost_of_equity = 0.10 # required return
growth_rate = 0.04    # dividend growth rate

stock_price = perpetuity_pv(dividend, cost_of_equity, growth_rate)
print(f"Gordon Growth Model Stock Valuation")
print(f"  Next dividend: ${dividend:.2f}")
print(f"  Required return: {cost_of_equity:.0%}")
print(f"  Growth rate: {growth_rate:.0%}")
print(f"  Fair price: ${stock_price:.2f}")

# ── Verify: perpetuity as limit of long annuity
C_perp, r_perp = 100, 0.08
pv_perp_exact = perpetuity_pv(C_perp, r_perp)
pv_annuity_long = annuity_pv(C_perp, r_perp, 1000)
print(f"\nPerpetuity PV (exact): ${pv_perp_exact:>10,.2f}")
print(f"Annuity PV (n=1000):   ${pv_annuity_long:>10,.2f}")
print(f"Difference:            ${abs(pv_perp_exact - pv_annuity_long):>10.6f}")
```

**Key takeaway from the verification:** A 1,000-year annuity has virtually the same PV as a perpetuity. This makes intuitive sense -- a dollar received 1,000 years from now is worth essentially zero today, so adding more payments beyond that doesn't change the total.

### Sensitivity to the Growth Rate

The Gordon Growth Model is extremely sensitive to the growth rate assumption. Small changes in $g$ can cause large swings in the estimated stock price. This is both its power and its danger -- let's see this in action.**Key takeaway from the verification:** A 1,000-year annuity has virtually the same PV as a perpetuity. This isn't a coincidence — it's a fundamental property of compound discounting.

### Why long annuities approximate perpetuities

The PV of an annuity is:

$$\text{PV}_{\text{annuity}} = C \times \frac{1 - (1+r)^{-n}}{r}$$

As $n \to \infty$, $(1+r)^{-n} \to 0$, leaving:

$$\text{PV}_{\text{perpetuity}} = \frac{C}{r}$$

For practical purposes, "infinity" arrives surprisingly quickly. A 50-year annuity at 5% captures over 91% of the perpetuity's value. A 100-year annuity captures over 99%. By 200 years, the difference is rounding error.

### The Gordon Growth Model

Perpetuities with growth — the **Gordon Growth Model** — are foundational to equity valuation:

$$P = \frac{D_1}{r - g}$$

where $P$ is stock price, $D_1$ is next year's dividend, $r$ is the required return, and $g$ is the perpetual growth rate. This formula values a stock as a growing perpetuity of dividends.

The model's elegance is also its danger: small changes in $r$ or $g$ produce huge changes in $P$. If $r = 10\%$ and $g = 5\%$, a \$1 dividend implies a \$20 stock price. If $g$ increases to 6%, the price jumps to \$25 — a 25% increase from a 1-percentage-point change.

> **Common Mistake:** The Gordon Growth Model requires $r > g$. If $g \geq r$, the formula produces a negative or infinite value — economic nonsense. This constraint reminds us that perpetual growth above the discount rate is impossible in equilibrium.

```python
# ── Sensitivity analysis: stock price vs growth rate
g_values = np.linspace(0.00, 0.08, 50)
prices = [perpetuity_pv(dividend, cost_of_equity, g) for g in g_values]

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(g_values * 100, prices, color=PRIMARY, linewidth=2)
ax.axhline(stock_price, color=SECONDARY, linestyle='--', alpha=0.7,
           label=f'g = {growth_rate:.0%}: ${stock_price:.2f}')
ax.set_xlabel('Dividend Growth Rate (%)')
ax.set_ylabel('Stock Price ($)')
ax.set_title('Gordon Growth Model: Price Sensitivity to Growth Rate')
ax.legend()
plt.tight_layout()
plt.show()
```

**Interpreting the sensitivity chart:** The stock price curve is not linear -- it rises steeply as $g$ approaches $r$ (10%). At $g = 0\%$, the stock is worth $30. At $g = 8\%$, it's worth $150 -- **five times** as much! And as $g$ approaches 10%, the price would go to infinity (which is why $g < r$ is required).

This extreme sensitivity is why analysts must be very careful about their growth rate assumptions. A seemingly small disagreement about whether a company will grow at 3% vs 5% translates into a massive difference in valuation.**Interpreting the sensitivity chart:** The stock price curve is not linear — it rises steeply as the growth rate approaches the discount rate. This is the "denominator effect" at work: as $r - g$ shrinks, $\frac{D_1}{r-g}$ explodes.

### The valuation knife-edge

The Gordon Growth Model's sensitivity to growth assumptions is its most important practical limitation. A 1-percentage-point change in $g$ can change the implied stock price by 20-50%, depending on starting parameters.

This is why responsible analysts:

1. **Use multiple valuation methods** — DCF, multiples, asset-based — and triangulate
2. **Conduct sensitivity analysis** — show price under a range of growth assumptions
3. **Stress-test the inputs** — what growth rate makes the current market price consistent with the model?

### When the model breaks down

Gordon Growth assumes a single perpetual growth rate. Real companies grow in stages:
* High growth (years 1-5): rapid expansion
* Moderate growth (years 6-10): maturing into industry growth rate
* Steady-state (year 11+): perpetual growth at GDP-like rate

The two-stage and three-stage **DDM (Dividend Discount Models)** address this by combining:
1. Explicit period of high growth (modelled year-by-year)
2. Terminal value computed via Gordon Growth at the steady-state rate

This more realistic structure is the foundation of modern equity valuation.

> **CFA Exam Tip:** The CFA curriculum tests Gordon Growth and its multi-stage extensions extensively. The key insight: the *terminal value* (the perpetuity portion) typically dominates the total valuation. If the terminal value represents 70%+ of total value (common), the model is essentially a perpetuity calculation with a few years of explicit forecasting bolted on.

## 6. Loan Amortization

### How Loans Work: The Big Picture

When you take out a mortgage or car loan, you make the same payment every month. But have you ever wondered: if the payment is the same each month, how does the loan actually get paid off?

The answer is that each payment is split into two parts:
1. **Interest** -- what you owe the bank for using their money
2. **Principal repayment** -- what actually reduces your loan balance

Here's the crucial insight: **early payments are mostly interest, while later payments are mostly principal.** This is because interest is calculated on the *remaining balance*, which starts high and gradually decreases.

### Calculating the Payment

A fully amortizing loan is really just an **ordinary annuity** from the bank's perspective. The bank gives you $PV$ today and receives $n$ equal payments of $PMT$. So the payment is just the annuity PV formula solved for $C$:

$$PMT = PV \cdot \frac{r}{1 - (1+r)^{-n}}$$

where:
- $PV$ = loan amount (principal)
- $r$ = interest rate per period
- $n$ = total number of payments

### Building the Amortization Schedule

For each period $t$:
1. **Interest portion:** $I_t = \text{Balance}_{t-1} \times r$ (interest on remaining balance)
2. **Principal portion:** $P_t = PMT - I_t$ (whatever's left goes to principal)
3. **New balance:** $\text{Balance}_t = \text{Balance}_{t-1} - P_t$

**Worked Example:** You borrow **$200,000** at **6% annually** for **30 years** (monthly payments).

- Monthly rate: $r = 0.06 / 12 = 0.005$
- Number of payments: $n = 30 \times 12 = 360$
- Monthly payment: $PMT = 200{,}000 \times \frac{0.005}{1 - (1.005)^{-360}} = \$1{,}199.10$

**First month breakdown:**
- Interest: $200,000 \times 0.005 = \$1{,}000.00$
- Principal: $1,199.10 - 1,000.00 = \$199.10$
- Remaining balance: $200,000 - 199.10 = \$199{,}800.90$

Notice that in the first month, only **$199 out of $1,199** actually goes toward paying off the loan! The rest is interest.

> **Key Concept:** Amortization is the process of gradually paying off a loan through equal periodic payments. The interest portion decreases and the principal portion increases over time because interest is calculated on the declining balance.

> **Common Mistake:** Thinking that if you're 15 years into a 30-year mortgage, you've paid off half the principal. You haven't -- because of front-loaded interest, you've probably paid off only about one-third of the principal by the halfway point.

> **CFA Exam Tip:** On the CFA exam, you might be asked to find the interest or principal portion of a specific payment. Remember: (1) compute the payment using the annuity formula, (2) interest = remaining balance times rate, (3) principal = payment minus interest.

Let's build a full amortization schedule and visualize the interest-principal breakdown.

```python
def loan_payment(pv, r, n):
    """Compute the periodic payment for a fully amortizing loan."""
    return pv * r / (1 - (1 + r) ** (-n))


def amortization_schedule(pv, r, n):
    """Build a full amortization schedule.
    
    Returns list of dicts with keys:
        period, payment, interest, principal, balance
    """
    pmt = loan_payment(pv, r, n)
    balance = pv
    schedule = []
    
    for t in range(1, n + 1):
        interest = balance * r
        principal = pmt - interest
        balance -= principal
        schedule.append({
            'period': t,
            'payment': pmt,
            'interest': interest,
            'principal': principal,
            'balance': max(balance, 0)
        })
    return schedule


# ── Example: $200,000 mortgage, 6% annual rate, 30-year term (monthly)
loan_amount = 200_000
annual_rate = 0.06
monthly_rate = annual_rate / 12
n_months = 30 * 12

pmt = loan_payment(loan_amount, monthly_rate, n_months)
print(f"Loan: ${loan_amount:,.0f} at {annual_rate:.0%} for 30 years")
print(f"Monthly payment: ${pmt:,.2f}")
print(f"Total payments:  ${pmt * n_months:,.2f}")
print(f"Total interest:  ${pmt * n_months - loan_amount:,.2f}")

# ── Full schedule
schedule = amortization_schedule(loan_amount, monthly_rate, n_months)

# Print first and last 5 rows
print(f"\n{'Period':>6} {'Payment':>10} {'Interest':>10} {'Principal':>10} {'Balance':>12}")
print("-" * 52)
for row in schedule[:5] + [{'period': '...', 'payment': 0, 'interest': 0, 'principal': 0, 'balance': 0}] + schedule[-5:]:
    if row['period'] == '...':
        print(f"{'...':>6}")
    else:
        print(f"{row['period']:>6} {row['payment']:>10,.2f} {row['interest']:>10,.2f} "
              f"{row['principal']:>10,.2f} {row['balance']:>12,.2f}")
```

**Interpreting the amortization schedule:**

Look at the key numbers:
- **Monthly payment:** about $1,199 -- you pay this every month for 30 years
- **Total payments:** about $431,677 over the life of the loan
- **Total interest:** about $231,677 -- you pay **more than the loan amount** in interest!

This is a sobering reality of long-term borrowing. On a 30-year mortgage, you typically pay more in interest than the house originally cost.

Now look at the first few rows: in month 1, $1,000 goes to interest and only $199 to principal. By the last few months, nearly the entire payment goes to principal.

The following visualization makes this shift crystal clear.**Interpreting the amortization schedule:**

The amortization schedule reveals the surprising mathematics of loan repayment:

1. **Early payments are mostly interest:** In the first month of a 30-year mortgage at 6%, approximately 83% of the payment goes to interest, only 17% to principal. The bank gets paid first; the principal slowly chips away.

2. **The interest-to-principal ratio shifts gradually:** As the outstanding balance shrinks, interest charges shrink too. By year 15, the split is roughly 50/50. By year 25, it's mostly principal.

3. **The total interest paid is enormous:** A 30-year, \$300,000 mortgage at 6% has total payments of approximately \$647,500 — meaning \$347,500 in interest alone. Total interest exceeds the original principal.

### Why the early-heavy interest pattern

Each month, you pay interest on the *remaining balance*. The remaining balance starts at the full loan amount and decreases each month. Therefore:

* Month 1 interest = full loan × monthly rate (largest)
* Month 360 interest = tiny remaining balance × monthly rate (smallest)

The level monthly payment is constructed so that early payments cover the high interest plus a small principal repayment, gradually shifting toward more principal and less interest over time.

### The 15-year vs 30-year choice

This pattern explains why 15-year mortgages save so much interest — not because the rate is lower (it usually is, but only slightly), but because the principal is paid down much faster, reducing the cumulative interest exposure.

| Mortgage | Rate | Monthly Payment | Total Interest |
|----------|------|-----------------|----------------|
| \$300K, 30-year | 6.0% | \$1,799 | \$347,514 |
| \$300K, 15-year | 5.5% | \$2,452 | \$141,290 |

The 15-year option saves over \$200K in interest but requires \$650 more per month — a trade-off between cash flow and total cost.

> **Key Concept:** Loan amortization is just the annuity formula in reverse. The bank lends a present value (the loan amount) and receives a stream of equal monthly payments (the annuity). The math is identical to computing the periodic payment from a known PV — just applied from the lender's perspective.

```python
# ── Visualization: Principal vs Interest over time
periods = [s['period'] for s in schedule]
interests = [s['interest'] for s in schedule]
principals = [s['principal'] for s in schedule]
balances = [s['balance'] for s in schedule]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: stacked area of interest vs principal
axes[0].stackplot(periods, interests, principals, 
                  labels=['Interest', 'Principal'],
                  colors=[SECONDARY, PRIMARY], alpha=0.8)
axes[0].set_xlabel('Month')
axes[0].set_ylabel('Amount ($)')
axes[0].set_title('Payment Breakdown: Interest vs Principal')
axes[0].legend(loc='center right')

# Right: remaining balance
axes[1].plot(periods, balances, color=PRIMARY, linewidth=2)
axes[1].fill_between(periods, 0, balances, alpha=0.15, color=PRIMARY)
axes[1].set_xlabel('Month')
axes[1].set_ylabel('Remaining Balance ($)')
axes[1].set_title('Loan Balance Over Time')

plt.tight_layout()
plt.show()
```

**Reading the charts:**

- **Left chart:** The total height of the stacked area is always the same ($1,199 per month). But the orange portion (interest) shrinks over time while the blue portion (principal) grows. The crossover point -- where you start paying more principal than interest -- happens roughly **halfway** through the loan term.

- **Right chart:** The loan balance curve is not linear -- it's concave. The balance drops slowly at first (because most of your payment is going to interest) and then accelerates toward the end. This is the visual signature of amortization.

### Practical Insight: Extra Payments

This chart explains why making **extra principal payments** early in a mortgage is so powerful. In year 1, only $199 per month goes to principal. If you paid an extra $200 per month, you would essentially **double** the rate at which you're paying down the loan, saving tens of thousands of dollars in interest over the life of the mortgage.**Reading the charts:**

The visualisation shows two related views of the amortization process:

1. **Stacked bar chart:** Each bar represents one period's payment, split between principal (lower portion) and interest (upper portion). The principal portion grows over time; the interest portion shrinks. The total bar height — the monthly payment — stays constant.

2. **Cumulative line chart:** The running total of principal paid vs interest paid. The cumulative interest line starts steep and flattens; the cumulative principal line starts flat and steepens. They cross at roughly the midpoint of the loan term — this is the year you've paid more principal than interest.

### Strategic implications

These charts have practical implications for borrowers:

* **Selling a home in the first 5-7 years** means you've barely paid down the principal — most equity comes from price appreciation, not loan paydown.
* **Refinancing in the early years** can dramatically reset the amortization clock — a 5-year-old 30-year mortgage refinanced into another 30-year is now a 35-year loan in total.
* **Extra principal payments early in the loan** save vastly more interest than the same payments late in the loan, because each early payment reduces the balance for many remaining periods.

### A common refinancing trap

Refinancing extends the loan term and resets the amortization. Someone 10 years into a 30-year mortgage who refinances into a new 30-year loan has effectively converted a 20-year remaining loan into a 30-year loan. Even at a lower rate, total interest paid often increases.

The right comparison: refinance into a *15-year* loan to actually save total interest, not a fresh 30-year loan that resets the schedule.

> **Common Mistake:** Borrowers often celebrate "saving \$200/month" through refinancing without computing the total cost. The lower payment may simply reflect a longer term. The proper analysis is total interest paid (or net present value of all payments) — not the monthly payment alone.

## 7. Applications: NPV and IRR

### Making Investment Decisions

Everything we've learned so far comes together in the two most important tools for investment decision-making: **Net Present Value (NPV)** and **Internal Rate of Return (IRR)**.

### Net Present Value (NPV)

**The concept:** NPV answers the question: *"Is this investment worth more than it costs?"*

You compute the present value of all future benefits, subtract the cost (investment), and see if there's value left over.

$$\text{NPV} = \sum_{t=0}^{n} \frac{CF_t}{(1+r)^t}$$

where:
- $CF_t$ = cash flow at time $t$ (usually $CF_0 < 0$ since it's the initial investment)
- $r$ = discount rate (usually the firm's cost of capital)

**Decision rule:**
- **NPV > 0:** Accept the project -- it creates value (you're getting more than your money's worth)
- **NPV < 0:** Reject the project -- it destroys value
- **NPV = 0:** Indifferent -- the project earns exactly the required return

**Worked Example:** A factory costs **$100,000** today and produces cash flows of $30,000, $35,000, $40,000, and $45,000 over the next 4 years. If the cost of capital is **10%**:

$$NPV = -100{,}000 + \frac{30{,}000}{1.10} + \frac{35{,}000}{1.10^2} + \frac{40{,}000}{1.10^3} + \frac{45{,}000}{1.10^4}$$

$$= -100{,}000 + 27{,}273 + 28{,}926 + 30{,}053 + 30{,}735 = \$16{,}987$$

Since NPV is positive, this project **creates $16,987 of value** beyond the required return. Accept it!

### Internal Rate of Return (IRR)

**The concept:** The IRR is the discount rate that makes the NPV exactly zero. Think of it as the project's "break-even" rate of return.

$$\sum_{t=0}^{n} \frac{CF_t}{(1+r^*)^t} = 0$$

**Decision rule:** Accept if IRR > cost of capital (the project earns more than it costs to fund).

There is no closed-form solution for IRR in general, so we solve numerically. We'll use the **bisection method** -- a simple algorithm that narrows down the answer by repeatedly halving an interval.

### NPV vs IRR: When Do They Disagree?

| Criterion | NPV | IRR |
|---|---|---|
| Question answered | "How much value does this create?" | "What rate of return does this earn?" |
| Accept/reject | NPV > 0 | IRR > cost of capital |
| Mutually exclusive projects | Always reliable | Can give wrong ranking |
| Preferred by CFA curriculum | Yes | Use with caution |

> **Key Concept:** For accept/reject decisions on independent projects, NPV and IRR always give the same answer. But for **ranking** mutually exclusive projects (where you can only pick one), NPV is superior because it measures absolute value creation.

> **CFA Exam Tip:** When NPV and IRR give conflicting rankings for mutually exclusive projects, **always go with NPV**. The CFA curriculum is clear on this preference.

Let's compare two projects.

```python
def npv(cashflows, r):
    """Compute Net Present Value.
    
    Parameters
    ----------
    cashflows : array-like -- cash flows starting at t=0
    r         : float     -- discount rate per period
    """
    return sum(cf / (1 + r) ** t for t, cf in enumerate(cashflows))


def irr_bisection(cashflows, tol=1e-10, max_iter=1000):
    """Compute IRR using bisection method.
    
    Searches for the rate r in [-0.5, 10.0] where NPV(r) = 0.
    """
    lo, hi = -0.5, 10.0
    
    if npv(cashflows, lo) * npv(cashflows, hi) > 0:
        raise ValueError("Cannot bracket IRR in the given range")
    
    for _ in range(max_iter):
        mid = (lo + hi) / 2
        val = npv(cashflows, mid)
        
        if abs(val) < tol:
            return mid
        
        if npv(cashflows, lo) * val < 0:
            hi = mid
        else:
            lo = mid
    
    return (lo + hi) / 2


# ── Compare two investment projects
project_A = [-100_000, 30_000, 35_000, 40_000, 45_000]
project_B = [-100_000, 10_000, 20_000, 50_000, 80_000]

discount_rate = 0.10

npv_A = npv(project_A, discount_rate)
npv_B = npv(project_B, discount_rate)
irr_A = irr_bisection(project_A)
irr_B = irr_bisection(project_B)

print(f"{'Metric':<20} {'Project A':>12} {'Project B':>12}")
print("-" * 46)
print(f"{'NPV @ 10%':<20} ${npv_A:>11,.2f} ${npv_B:>11,.2f}")
print(f"{'IRR':<20} {irr_A:>11.2%} {irr_B:>11.2%}")
print(f"\nDecision at r=10%: {'Project A' if npv_A > npv_B else 'Project B'} has higher NPV")
```

**Interpreting the results:**

Both projects cost $100,000, but they have very different cash flow timing:
- **Project A** generates cash sooner (front-loaded cash flows)
- **Project B** generates more cash later (back-loaded cash flows)

At a 10% discount rate, look at what happens: Project A has higher NPV *and* higher IRR. Both methods agree here.

But what if we changed the discount rate? At very low discount rates, Project B (with its larger total cash flows) might have a higher NPV. This is the classic scenario where NPV and IRR can give different rankings.

The NPV profile chart below shows this beautifully -- it plots NPV as a function of the discount rate.**Interpreting the results:**

The NPV calculation aggregates all the cash flows of a project — initial investment (negative), future cash inflows (positive) — discounted to today. The decision rule is simple:

* **NPV > 0:** Accept the project. It earns more than the required return.
* **NPV < 0:** Reject the project. It earns less than the required return.
* **NPV = 0:** Indifferent. The project earns exactly the required return.

The required return (discount rate) is the threshold: it's the minimum return demanded for taking the project's risk.

### The economic meaning of NPV

NPV is *not* the project's profit. It's the **excess value created** above the opportunity cost of capital. A project with NPV = \$1M means: after compensating investors for their required return, this project creates \$1M of additional value.

This is why NPV is the theoretically correct decision rule: it directly measures wealth creation. A project with positive NPV makes shareholders richer; one with negative NPV makes them poorer.

### IRR vs NPV — when they conflict

The Internal Rate of Return (IRR) is the discount rate that makes NPV = 0. Both NPV and IRR usually agree on accept/reject decisions, but they can conflict when:

1. **Comparing mutually exclusive projects:** NPV correctly identifies the larger value creator; IRR may favour smaller-scale projects with higher returns.

2. **Non-conventional cash flows:** Projects with multiple sign changes can have *multiple IRRs*, making the metric ambiguous. NPV is always well-defined.

3. **Different scale or duration:** Comparing a small short project (high IRR) to a large long project (large NPV) can favour different choices depending on the metric.

> **CFA Exam Tip:** When NPV and IRR disagree, **always trust NPV**. NPV is the theoretically correct decision rule because it directly measures shareholder wealth creation. IRR is a useful summary statistic but can produce misleading rankings, particularly for projects with different scales, durations, or cash flow patterns.

```python
# ── NPV Profile: NPV as a function of discount rate
rates_range = np.linspace(0.0, 0.35, 200)
npvs_A = [npv(project_A, r) for r in rates_range]
npvs_B = [npv(project_B, r) for r in rates_range]

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(rates_range * 100, npvs_A, color=PRIMARY, linewidth=2, label='Project A')
ax.plot(rates_range * 100, npvs_B, color=SECONDARY, linewidth=2, label='Project B')
ax.axhline(0, color='black', linewidth=0.8)
ax.axvline(irr_A * 100, color=PRIMARY, linestyle=':', alpha=0.7, label=f'IRR(A) = {irr_A:.1%}')
ax.axvline(irr_B * 100, color=SECONDARY, linestyle=':', alpha=0.7, label=f'IRR(B) = {irr_B:.1%}')
ax.set_xlabel('Discount Rate (%)')
ax.set_ylabel('NPV ($)')
ax.set_title('NPV Profile: Comparing Two Projects')
ax.legend()
plt.tight_layout()
plt.show()
```

**Reading the NPV profile:**

- Each line shows a project's NPV at different discount rates
- The **x-intercept** of each line is that project's **IRR** (where NPV = 0)
- At **low discount rates**, the lines may cross -- meaning the ranking flips
- At the **crossover rate**, both projects have equal NPV

This chart is one of the most important in corporate finance. It shows exactly when and why NPV and IRR can give different rankings.

> **Common Mistake:** Some students think higher IRR always means a better project. It doesn't! A small project with 50% IRR might create less value than a large project with 15% IRR. NPV measures absolute value creation; IRR measures the rate of return. For mutually exclusive projects, value creation (NPV) is what matters.**Reading the NPV profile:**

The NPV profile plots NPV on the y-axis against discount rate on the x-axis. Several key features:

1. **The curve slopes downward:** Higher discount rates make future cash flows worth less in present terms, reducing NPV. For typical projects (positive future cash flows), the relationship is monotonically decreasing.

2. **The IRR is the x-intercept:** Where the NPV curve crosses zero is the project's IRR — the discount rate at which the project breaks even.

3. **The y-intercept is the undiscounted NPV:** At a 0% discount rate, NPV equals the simple sum of all cash flows.

4. **The curve is convex:** As the discount rate rises, NPV falls less steeply. This is again the exponential-decay nature of discounting.

### Using NPV profiles for decision-making

NPV profiles are particularly useful for comparing two projects:

* The crossover point — where two NPV profiles intersect — is the discount rate at which the projects produce equal NPVs.
* Below the crossover: one project is preferred. Above: the other is preferred.
* This "switching" reveals the role of the cost of capital in project selection.

### Multiple IRRs — a warning

Projects with **non-conventional cash flows** (multiple sign changes — e.g., outflow, inflow, outflow) can have *multiple* IRRs. The NPV profile crosses zero more than once. In these cases:

* IRR is ambiguous — which one do you use?
* NPV is unambiguous — there's still one NPV at any given discount rate
* The accept/reject decision based on NPV is always reliable

This is another reason NPV is preferred to IRR for capital budgeting decisions.

> **Key Concept:** The NPV profile is one of the most powerful diagnostic tools in capital budgeting. By plotting NPV as a function of the discount rate, you can simultaneously see: (1) the project's IRR, (2) the sensitivity of value to assumptions about the cost of capital, (3) the project's NPV at any specific rate of interest. A single chart contains the answers to many "what if" questions.

## Summary: Key Formulas at a Glance

| Concept | Formula | When to Use |
|---|---|---|
| Future Value | $FV = PV(1+r)^n$ | "How much will my investment grow to?" |
| Present Value | $PV = FV/(1+r)^n$ | "What is a future payment worth today?" |
| EAR | $(1 + r_s/m)^m - 1$ | Comparing rates with different compounding |
| Annuity PV | $C \cdot \frac{1-(1+r)^{-n}}{r}$ | Equal payments over a fixed period |
| Perpetuity PV | $C/r$ | Equal payments forever |
| Growing Perpetuity | $C/(r-g)$ | Growing payments forever (Gordon Model) |
| Loan Payment | $PV \cdot \frac{r}{1-(1+r)^{-n}}$ | Amortizing loan payments |
| NPV | $\sum CF_t/(1+r)^t$ | Project valuation |

> **Key Concept:** All of these formulas are variations on one simple idea: a dollar today is worth more than a dollar tomorrow. Everything in TVM is built by applying this principle to different patterns of cash flows.
