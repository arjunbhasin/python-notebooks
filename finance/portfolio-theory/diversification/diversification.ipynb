## The Only Free Lunch in Finance

> **Key Concept:** Diversification is often called the "only free lunch in finance." By combining assets that don't move in perfect lockstep, you can reduce portfolio risk without sacrificing expected return. This is not a trick or an approximation --- it is a mathematical certainty.

### Why Does Diversification Work?

The intuition is simple: when one asset zigs, another may zag. If you hold both, the zigs and zags partially cancel out, leaving you with a smoother ride.

Imagine you own an umbrella company and a sunscreen company. On rainy days the umbrella company does well and the sunscreen company suffers. On sunny days the reverse happens. Individually, each company is risky --- its revenue swings with the weather. But if you own *both*, your combined revenue is stable regardless of the weather. You have diversified away the "weather risk."

This is not just a cute story. It is a precise mathematical result, and the rest of this notebook will prove it from first principles.

**A Simple Example:**

Suppose you have two stocks, each with:
- Expected return: 10%
- Standard deviation: 20%
- Correlation: 0.3

An equal-weight (50/50) portfolio has:
- Expected return: $0.5 \times 10\% + 0.5 \times 10\% = 10\%$ (same as individual stocks!)
- Portfolio variance: $0.5^2(0.20^2) + 0.5^2(0.20^2) + 2(0.5)(0.5)(0.3)(0.20)(0.20) = 0.026$
- Portfolio std dev: $\sqrt{0.026} = 16.1\%$ (LOWER than individual stocks!)

**You got the same return with less risk.** That's the free lunch.

Let's walk through that variance calculation step by step, because this formula is the foundation of everything that follows:

| Term | Calculation | Value |
|------|-------------|-------|
| $w_1^2 \sigma_1^2$ | $0.5^2 \times 0.20^2$ | 0.0100 |
| $w_2^2 \sigma_2^2$ | $0.5^2 \times 0.20^2$ | 0.0100 |
| $2 w_1 w_2 \rho \sigma_1 \sigma_2$ | $2 \times 0.5 \times 0.5 \times 0.3 \times 0.20 \times 0.20$ | 0.0060 |
| **Portfolio variance** | Sum of above | **0.0260** |
| **Portfolio std dev** | $\sqrt{0.0260}$ | **16.1%** |

Notice that if the correlation were 1.0 instead of 0.3, the cross-term would be $2 \times 0.5 \times 0.5 \times 1.0 \times 0.20 \times 0.20 = 0.0200$, giving a portfolio variance of $0.0100 + 0.0100 + 0.0200 = 0.0400$ and a portfolio std dev of $\sqrt{0.04} = 20\%$ --- no benefit at all. **The diversification benefit comes entirely from the correlation being less than 1.**

> **CFA Exam Tip:** On the CFA exam, you will often be asked to calculate the portfolio standard deviation for a two-asset portfolio. The formula is $\sigma_p = \sqrt{w_1^2 \sigma_1^2 + w_2^2 \sigma_2^2 + 2 w_1 w_2 \rho_{12} \sigma_1 \sigma_2}$. Memorise it cold. Most questions test whether you can correctly identify $\rho$ (correlation) vs $\text{Cov}$ (covariance). Remember: $\text{Cov}(R_1, R_2) = \rho_{12} \sigma_1 \sigma_2$.

### The Mathematics of Diversification

For an equal-weight portfolio of $N$ assets, each with variance $\sigma^2$ and pairwise correlation $\rho$:

$$\sigma_p^2 = \frac{\sigma^2}{N} + \frac{N-1}{N}\rho\sigma^2$$

As $N \to \infty$:

$$\sigma_p^2 \to \rho\sigma^2$$

**In plain English:** The portfolio variance has two pieces. The first piece, $\frac{\sigma^2}{N}$, is the average variance divided by the number of assets --- it shrinks toward zero as you add more assets. This is the **diversifiable** (idiosyncratic, firm-specific) risk. The second piece, $\frac{N-1}{N}\rho\sigma^2$, approaches $\rho\sigma^2$ as $N$ grows. This is the **non-diversifiable** (systematic, market) risk.

**Where does this formula come from?** Consider an equal-weight portfolio with $w_i = 1/N$ for all $i$. The portfolio variance is:

$$\sigma_p^2 = \sum_{i=1}^{N} w_i^2 \sigma_i^2 + \sum_{i \neq j} w_i w_j \text{Cov}(R_i, R_j)$$

Since all assets have the same variance $\sigma^2$ and all pairs have the same correlation $\rho$:

$$\sigma_p^2 = N \cdot \frac{1}{N^2} \cdot \sigma^2 + N(N-1) \cdot \frac{1}{N^2} \cdot \rho \sigma^2 = \frac{\sigma^2}{N} + \frac{N-1}{N}\rho\sigma^2$$

There are $N$ variance terms (the diagonal of the covariance matrix) and $N(N-1)$ covariance terms (the off-diagonal entries).

**Worked example with numbers:**

Suppose each asset has $\sigma = 20\%$ and $\rho = 0.3$.

| $N$ | $\frac{\sigma^2}{N}$ (diversifiable) | $\frac{N-1}{N}\rho\sigma^2$ (systematic) | $\sigma_p^2$ | $\sigma_p$ |
|-----|--------------------------------------|-------------------------------------------|-------------|------------|
| 1 | 0.0400 | 0.0000 | 0.0400 | 20.00% |
| 2 | 0.0200 | 0.0060 | 0.0260 | 16.12% |
| 5 | 0.0080 | 0.0096 | 0.0176 | 13.27% |
| 10 | 0.0040 | 0.0108 | 0.0148 | 12.17% |
| 30 | 0.0013 | 0.0116 | 0.0129 | 11.37% |
| 100 | 0.0004 | 0.0119 | 0.0123 | 11.07% |
| $\infty$ | 0.0000 | 0.0120 | 0.0120 | 10.95% |

Notice how the diversifiable risk column shrinks from 0.04 to nearly zero, while the systematic risk column converges to $\rho \sigma^2 = 0.3 \times 0.04 = 0.012$. The floor of $\sqrt{0.012} = 10.95\%$ is the irreducible systematic risk.

> **Key Concept:** No amount of diversification can eliminate systematic risk ($\rho\sigma^2$). This is the risk of the market itself --- recessions, interest rate changes, pandemics --- events that affect all assets simultaneously.

> **Common Mistake:** Students sometimes think that with enough assets you can eliminate *all* portfolio risk. This is only true in the special case where $\rho = 0$ (assets are uncorrelated). In practice, most stocks have positive correlation with each other (they all respond to the same macroeconomic forces), so there is always a floor to how much risk you can eliminate.

## 3. Diversification Benefit --- Variance Reduction with N Assets

We now formalise the diversification result we previewed in the introduction. The question is: **how does portfolio risk change as we add more assets?**

### The Equal-Weight Portfolio Variance Formula

For an equally-weighted portfolio of $N$ assets, each with variance $\sigma^2$ and pairwise correlation $\rho$:

$$\sigma_P^2 = \frac{\sigma^2}{N} + \frac{N-1}{N} \rho \sigma^2$$

**Variable definitions:**

| Symbol | Meaning |
|--------|---------|
| $\sigma_P^2$ | Variance of the portfolio |
| $\sigma^2$ | Variance of each individual asset (assumed identical) |
| $\rho$ | Pairwise correlation between any two assets (assumed identical) |
| $N$ | Number of assets in the portfolio |

**In plain English:** The portfolio variance is a weighted sum of two components. The first component, $\frac{\sigma^2}{N}$, represents the average individual variance, which gets diluted as you add more assets. The second component, $\frac{N-1}{N}\rho\sigma^2$, captures the average covariance between assets, which does *not* get diluted --- it converges to $\rho\sigma^2$ as $N$ grows.

### Deriving the Formula

Start with the general portfolio variance formula:

$$\sigma_P^2 = \mathbf{w}^\top \Sigma \mathbf{w} = \sum_{i=1}^{N} \sum_{j=1}^{N} w_i w_j \text{Cov}(R_i, R_j)$$

For equal weights ($w_i = 1/N$) with identical variances and correlations:

$$\sigma_P^2 = \sum_{i=1}^{N} \frac{1}{N^2} \sigma^2 + \sum_{i \neq j} \frac{1}{N^2} \rho \sigma^2$$

There are $N$ diagonal terms ($i = j$) and $N(N-1)$ off-diagonal terms ($i \neq j$), so:

$$\sigma_P^2 = \frac{N}{N^2} \sigma^2 + \frac{N(N-1)}{N^2} \rho \sigma^2 = \frac{\sigma^2}{N} + \frac{N-1}{N} \rho \sigma^2$$

### The Limiting Behaviour

As $N \to \infty$: $\sigma_P^2 \to \rho \sigma^2$ (only systematic risk remains).

This limit tells us the **floor** for portfolio risk. No matter how many assets you add, you cannot reduce the portfolio standard deviation below $\sqrt{\rho} \cdot \sigma$.

**Worked example:** With $\sigma = 20\%$ and $\rho = 0.25$:

$$\sigma_{P,\text{floor}} = \sqrt{0.25 \times 0.04} = \sqrt{0.01} = 10\%$$

So even with infinitely many assets, you cannot push portfolio volatility below 10%. The reduction from 20% to 10% is substantial, but the remaining 10% is systematic risk that no amount of diversification can eliminate.

> **CFA Exam Tip:** The exam often asks "what is the minimum portfolio standard deviation achievable through diversification?" The answer is $\sigma\sqrt{\rho}$ --- the square root of the average covariance. Make sure you can derive this quickly.

### Simulating the Diversification Effect

Let's verify this mathematically. We'll create a portfolio of $N$ stocks, each with the same individual volatility but imperfect correlation, and watch what happens to portfolio risk as $N$ grows.

The code below implements the equal-weight portfolio variance formula directly and plots the portfolio volatility for four different correlation levels ($\rho = 0, 0.2, 0.5, 0.8$). Dashed lines show the asymptotic floors.

> **What to watch for:** The portfolio standard deviation should drop rapidly at first (adding a 2nd stock helps enormously) but then flatten out. The floor it approaches is the **systematic risk** --- the part you cannot diversify away.

**Predictions before you look at the chart:**

| $\rho$ | Asymptotic floor = $\sigma\sqrt{\rho}$ |
|--------|---------------------------------------|
| 0.0 | $0.20 \times \sqrt{0.0} = 0\%$ |
| 0.2 | $0.20 \times \sqrt{0.2} = 8.94\%$ |
| 0.5 | $0.20 \times \sqrt{0.5} = 14.14\%$ |
| 0.8 | $0.20 \times \sqrt{0.8} = 17.89\%$ |

When $\rho = 0$, the floor is zero --- you can diversify away *all* risk. When $\rho = 0.8$, the floor is close to the individual asset volatility --- diversification barely helps because the assets are so highly correlated.

Let's verify with code.

```python
def equal_weight_portfolio_var(sigma, rho, N):
    """Variance of equally-weighted portfolio."""
    return sigma**2 / N + (N - 1) / N * rho * sigma**2

sigma = 0.20  # individual asset vol
N_range = np.arange(1, 101)

fig, ax = plt.subplots()
for rho, c, ls in [(0.0, TERTIARY, '-'), (0.2, PRIMARY, '-'), (0.5, SECONDARY, '-'), (0.8, ACCENT, '-')]:
    port_vol = np.sqrt([equal_weight_portfolio_var(sigma, rho, n) for n in N_range])
    ax.plot(N_range, port_vol * 100, color=c, linewidth=2, label=f'\u03c1 = {rho}')
    # Asymptote
    ax.axhline(np.sqrt(rho) * sigma * 100, color=c, linestyle='--', alpha=0.3)

ax.set_xlabel('Number of Assets (N)')
ax.set_ylabel('Portfolio Volatility (%)')
ax.set_title(f'Diversification Effect (individual \u03c3 = {sigma*100:.0f}%)')
ax.legend()
ax.set_xlim(1, 100)
plt.tight_layout()
plt.show()
```

### Interpreting the Diversification Curve

The plot above confirms our theoretical predictions exactly:

- **Green line ($\rho = 0$):** Portfolio volatility drops toward zero as $N$ increases. With uncorrelated assets, all risk is idiosyncratic and can be fully diversified away. This is the theoretical ideal but rarely occurs in practice --- most financial assets share at least some common risk factors.

- **Blue line ($\rho = 0.2$):** Rapid initial decline, levelling off around 9% --- close to the theoretical floor of $20\% \times \sqrt{0.2} = 8.94\%$. This is a realistic correlation for a broadly diversified portfolio across different sectors and asset classes.

- **Coral line ($\rho = 0.5$):** The floor is about 14%. With moderately correlated assets (e.g., stocks within the same sector), diversification helps but hits diminishing returns quickly.

- **Gold line ($\rho = 0.8$):** Diversification provides very little benefit. The floor is about 18%, barely below the individual asset volatility of 20%. This would be like holding multiple tech stocks that all move together.

> **Key Concept:** Diversification benefits are **asymptotic**. The first few assets you add provide enormous risk reduction. Beyond about 20-30 stocks, you're getting diminishing returns. The remaining risk is **systematic (market) risk** --- it cannot be diversified away no matter how many assets you hold.

**Practical implications:**

- **Index funds** with 500+ stocks aren't dramatically less risky than those with 50 stocks (the curve has already flattened)
- **Systematic risk** must be compensated by higher expected returns (this is the foundation of CAPM)
- **Idiosyncratic risk** is NOT compensated because investors can easily eliminate it by holding a diversified portfolio
- The biggest "bang for your buck" comes from the first 15-20 assets --- after that, you're mostly adding complexity without meaningfully reducing risk

> **CFA Exam Tip:** A common exam question is: "An investor holds a single stock. What type of risk can be reduced by adding more stocks?" The answer is **unsystematic (idiosyncratic, firm-specific, diversifiable) risk**. Systematic (market, non-diversifiable) risk cannot be reduced through diversification. The CFA curriculum uses all these synonyms --- make sure you recognise them all.

## 4. Correlation and Portfolio Risk

Diversification is most powerful when correlations are low or negative.

We have seen that correlation $\rho$ is the critical variable that determines how much diversification benefit you get. Let's now examine this relationship more carefully using a **two-asset portfolio**, which is the simplest and most instructive case.

### The Two-Asset Portfolio Variance Formula

For a portfolio of two assets with weights $w_1$ and $w_2 = 1 - w_1$:

$$\sigma_p^2 = w_1^2 \sigma_1^2 + w_2^2 \sigma_2^2 + 2 w_1 w_2 \rho_{12} \sigma_1 \sigma_2$$

**Variable definitions:**

| Symbol | Meaning |
|--------|---------|
| $w_1, w_2$ | Portfolio weights (fractions of wealth invested in each asset) |
| $\sigma_1, \sigma_2$ | Standard deviations of asset returns |
| $\rho_{12}$ | Correlation coefficient between the two assets' returns |

**In plain English:** Portfolio variance is the sum of three terms:
1. The variance contributed by Asset 1 alone: $w_1^2 \sigma_1^2$
2. The variance contributed by Asset 2 alone: $w_2^2 \sigma_2^2$
3. The **interaction term**: $2 w_1 w_2 \rho_{12} \sigma_1 \sigma_2$

The interaction term is the key to diversification. When $\rho_{12} < 1$, this term is smaller than it would be under perfect correlation, and the portfolio variance is less than the weighted average of individual variances.

### Three Special Cases

**Case 1: Perfect positive correlation ($\rho = +1$)**

$$\sigma_p = w_1 \sigma_1 + w_2 \sigma_2$$

Portfolio risk is simply the weighted average of individual risks. **No diversification benefit at all.** The portfolio lies on a straight line between the two assets in risk-return space.

**Case 2: Zero correlation ($\rho = 0$)**

$$\sigma_p = \sqrt{w_1^2 \sigma_1^2 + w_2^2 \sigma_2^2}$$

The cross-term vanishes entirely. Portfolio risk is significantly lower than the weighted average. For example, with $w_1 = w_2 = 0.5$ and $\sigma_1 = \sigma_2 = 20\%$: $\sigma_p = \sqrt{0.01 + 0.01} = 14.14\%$, a 29% reduction.

**Case 3: Perfect negative correlation ($\rho = -1$)**

$$\sigma_p = |w_1 \sigma_1 - w_2 \sigma_2|$$

You can completely eliminate all risk by choosing weights such that $w_1 \sigma_1 = w_2 \sigma_2$, i.e.:

$$w_1^* = \frac{\sigma_2}{\sigma_1 + \sigma_2}, \quad w_2^* = \frac{\sigma_1}{\sigma_1 + \sigma_2}$$

**Worked example:** If $\sigma_1 = 15\%$ and $\sigma_2 = 25\%$ with $\rho = -1$:

$$w_1^* = \frac{0.25}{0.15 + 0.25} = 62.5\%, \quad w_2^* = 37.5\%$$

Portfolio risk: $|0.625 \times 0.15 - 0.375 \times 0.25| = |0.09375 - 0.09375| = 0$. Zero risk!

> **Common Mistake:** Students sometimes assume that $\rho = -1$ means you should hold 50/50. But the zero-risk portfolio requires weights proportional to the *other* asset's volatility. With unequal volatilities, 50/50 does NOT eliminate all risk even at $\rho = -1$.

### Visualising the Correlation Effect

Correlation is the engine of diversification. Let's see what happens to a 2-asset portfolio as we vary the correlation from $-1$ to $+1$.

**What we expect:**
- At $\rho = +1$: No diversification benefit at all --- portfolio risk is the weighted average
- At $\rho = 0$: Significant risk reduction
- At $\rho = -1$: Perfect hedge --- we can eliminate all risk with the right weights

In the chart below, the **shaded area** between the portfolio volatility curve and the horizontal dashed line represents the **diversification benefit** --- the gap between what risk *would be* with perfect correlation and what it *actually is*. A larger shaded area means more diversification benefit.

We use $\sigma_1 = 15\%$, $\sigma_2 = 25\%$, and equal weights ($w_1 = w_2 = 0.5$). Let's see the result.

```python
# 2-asset case: portfolio vol as function of correlation
sig1, sig2 = 0.15, 0.25
w1_opt = 0.5
w2_opt = 1 - w1_opt

rhos = np.linspace(-1, 1, 200)
port_vols = np.sqrt(w1_opt**2 * sig1**2 + w2_opt**2 * sig2**2 + 2 * w1_opt * w2_opt * rhos * sig1 * sig2)

# Weighted average vol (no diversification benefit)
avg_vol = w1_opt * sig1 + w2_opt * sig2

fig, ax = plt.subplots()
ax.plot(rhos, port_vols * 100, color=PRIMARY, linewidth=2, label='Portfolio vol')
ax.axhline(avg_vol * 100, color=SECONDARY, linestyle='--', linewidth=2, label='Wtd avg vol (\u03c1=1)')
ax.fill_between(rhos, port_vols * 100, avg_vol * 100, alpha=0.15, color=TERTIARY)
ax.set_xlabel('Correlation (\u03c1)')
ax.set_ylabel('Portfolio Volatility (%)')
ax.set_title('Diversification Benefit (shaded area)')
ax.legend()
plt.tight_layout()
plt.show()

# Maximum diversification benefit at \u03c1 = -1
w_zero_vol = sig2 / (sig1 + sig2)
print(f"At \u03c1 = -1, zero-vol portfolio: w1 = {w_zero_vol:.2%}, w2 = {1-w_zero_vol:.2%}")
```

### What the Correlation Chart Tells Us

The chart confirms our theoretical predictions beautifully:

- **At $\rho = +1$ (right edge):** The blue curve meets the dashed coral line. Portfolio volatility equals the weighted average of individual volatilities ($0.5 \times 15\% + 0.5 \times 25\% = 20\%$). No diversification benefit.

- **At $\rho = 0$ (centre):** Portfolio volatility drops to $\sqrt{0.5^2 \times 0.15^2 + 0.5^2 \times 0.25^2} = \sqrt{0.005625 + 0.015625} = \sqrt{0.02125} \approx 14.58\%$. That's a meaningful reduction from 20%.

- **At $\rho = -1$ (left edge):** Portfolio volatility hits its minimum. With the right weights ($w_1 = 62.5\%$, $w_2 = 37.5\%$), you can reach zero volatility. With our 50/50 weights, we get $|0.5 \times 0.15 - 0.5 \times 0.25| \times 100 = 5\%$.

The printed output confirms: at $\rho = -1$, the zero-volatility portfolio requires $w_1 = 62.50\%$ and $w_2 = 37.50\%$, which matches our formula $w_1^* = \sigma_2 / (\sigma_1 + \sigma_2) = 0.25 / 0.40 = 62.5\%$.

Notice the dramatic effect of correlation:
- With $\rho = -0.5$, a 50/50 portfolio of two 20%-volatility assets has risk well below 20%
- The **minimum variance portfolio** shifts as correlation changes
- Only at $\rho = +1$ does "diversification" completely fail

> **CFA Exam Tip:** The portfolio variance formula for two assets is $\sigma_p^2 = w_1^2\sigma_1^2 + w_2^2\sigma_2^2 + 2w_1w_2\rho\sigma_1\sigma_2$. The key term is the last one --- it can be negative if $\rho < 0$, which is how diversification reduces risk. On the exam, you might be asked: "What is the maximum risk reduction achievable?" If $\rho = -1$, risk can be completely eliminated.

> **Common Mistake:** In practice, finding two assets with $\rho = -1$ is virtually impossible. Even stocks and bonds, the classic "diversifying" pair, typically have correlations between $-0.3$ and $+0.3$ depending on the economic regime. Never assume extreme negative correlations will persist.

## 5. Systematic vs Idiosyncratic Risk Decomposition

We have seen that diversification eliminates some risk but not all. Now we formalise this distinction using the **single-factor model**.

### The Single-Factor Model

The single-factor model assumes that every asset's return can be decomposed into two parts:

$$R_i = \alpha_i + \beta_i R_M + \epsilon_i$$

**Variable definitions:**

| Symbol | Meaning |
|--------|---------|
| $R_i$ | Return on asset $i$ |
| $\alpha_i$ | Asset-specific expected return (intercept) |
| $\beta_i$ | Sensitivity of asset $i$ to the market factor |
| $R_M$ | Return on the market portfolio |
| $\epsilon_i$ | Idiosyncratic (firm-specific) shock; $E[\epsilon_i] = 0$, $\text{Cov}(\epsilon_i, R_M) = 0$ |

**In plain English:** Every stock's return has two drivers: (1) what the market does (captured by $\beta_i R_M$), and (2) something unique to that stock (captured by $\epsilon_i$). Think of it this way: when Apple's stock moves, part of that move is because the whole market moved (S&P 500 went up 1%, and Apple with $\beta = 1.2$ went up 1.2% just from that). The rest is Apple-specific news --- a new product launch, an earnings surprise, a CEO change.

### Variance Decomposition

Taking the variance of both sides:

$$\sigma_i^2 = \beta_i^2 \sigma_M^2 + \sigma_{\epsilon_i}^2$$

Where:
- $\beta_i^2 \sigma_M^2$ = **systematic risk** (market risk) of asset $i$
- $\sigma_{\epsilon_i}^2$ = **idiosyncratic risk** (firm-specific risk) of asset $i$

**The key insight:** When you form a portfolio of $N$ assets, the idiosyncratic shocks $\epsilon_i$ are independent across assets. By the law of large numbers, their average converges to zero. But the systematic component $\beta_i R_M$ is common to all assets --- it does not average out. This is a rigorous statement of why diversification eliminates idiosyncratic risk but not systematic risk.

> **Key Concept:** This decomposition is the theoretical foundation of the **Capital Asset Pricing Model (CAPM)**. Since idiosyncratic risk can be diversified away for free, the market should not reward investors for bearing it. Only systematic risk (measured by $\beta$) should earn a risk premium. This is why $\beta$ is the single most important risk measure in modern finance.

**Worked example:** Suppose a stock has $\beta = 1.3$ and the market volatility is $\sigma_M = 15\%$. If the stock's total volatility is $\sigma_i = 25\%$, then:

- Systematic variance: $\beta_i^2 \sigma_M^2 = 1.3^2 \times 0.15^2 = 1.69 \times 0.0225 = 0.0380$
- Total variance: $\sigma_i^2 = 0.25^2 = 0.0625$
- Idiosyncratic variance: $\sigma_{\epsilon_i}^2 = 0.0625 - 0.0380 = 0.0245$
- Systematic risk as percentage of total: $0.0380 / 0.0625 = 60.8\%$

So about 61% of this stock's risk is systematic (non-diversifiable), and 39% is idiosyncratic (diversifiable). A well-diversified portfolio would eliminate that 39%.

> **CFA Exam Tip:** The fraction of variance explained by the market factor is called **$R^2$** (R-squared). It equals $\beta_i^2 \sigma_M^2 / \sigma_i^2$. A stock with high $R^2$ is mostly driven by the market; a stock with low $R^2$ is mostly driven by firm-specific factors. The CFA exam may ask you to compute the proportion of systematic vs unsystematic risk using $R^2$.

### Marginal Contribution to Risk (MCTR)

When managing a real portfolio, you need to know: **which position is contributing the most risk?**

This is not as simple as it sounds. An asset with high volatility might contribute less risk than you expect if it has low correlation with the rest of the portfolio. Conversely, a low-volatility asset that is highly correlated with everything else might be a bigger risk contributor than you think.

The **marginal contribution to risk** of asset $i$ is:

$$\text{MCTR}_i = \frac{\partial \sigma_p}{\partial w_i} = \frac{(\Sigma \mathbf{w})_i}{\sigma_p}$$

**In plain English:** MCTR answers the question: "If I increase the weight of asset $i$ by a tiny amount (say, from 20% to 20.01%), how much does portfolio risk change?" It is the rate of change of portfolio risk with respect to the weight of asset $i$.

**Why the formula works:** The numerator $(\Sigma \mathbf{w})_i$ is the $i$-th element of the vector you get by multiplying the covariance matrix by the weight vector. This captures how asset $i$'s returns covary with the overall portfolio. Dividing by $\sigma_p$ normalises to give a rate of change in volatility units.

### Component Contribution to Risk (CCTR)

The **component contribution to risk (CCTR)** is:

$$\text{CCTR}_i = w_i \times \text{MCTR}_i$$

**In plain English:** CCTR answers a different question: "How much of the portfolio's total risk is attributable to asset $i$?" It accounts for both the marginal risk of the asset and how much of it you hold.

The CCTRs have a beautiful property --- they sum to the total portfolio volatility:

$$\sum_i \text{CCTR}_i = \sigma_p$$

This is known as **Euler's decomposition** and follows from Euler's theorem on homogeneous functions (portfolio volatility is homogeneous of degree 1 in the weights).

### Percentage Risk Contribution

Dividing each CCTR by portfolio volatility gives the **percentage risk contribution**:

$$\text{PRC}_i = \frac{\text{CCTR}_i}{\sigma_p} = \frac{w_i \times \text{MCTR}_i}{\sigma_p}$$

These percentages sum to 100%. If asset $i$ has $\text{PRC}_i = 40\%$, it means 40% of the portfolio's total risk comes from asset $i$.

> **Key Concept:** An **Equal Risk Contribution (ERC)** portfolio sets all CCTRs equal. This is a popular "risk parity" approach --- every asset contributes the same amount of risk to the portfolio. We will construct one later.

**Worked example:** Suppose a 2-asset portfolio with $w = [0.6, 0.4]$, $\sigma = [15\%, 25\%]$, $\rho = 0.3$.

First, build the covariance matrix:
$$\Sigma = \begin{pmatrix} 0.0225 & 0.01125 \\ 0.01125 & 0.0625 \end{pmatrix}$$

Portfolio variance: $w^\top \Sigma w = 0.6^2(0.0225) + 0.4^2(0.0625) + 2(0.6)(0.4)(0.01125) = 0.0081 + 0.01 + 0.0054 = 0.0235$

Portfolio vol: $\sigma_p = \sqrt{0.0235} = 15.33\%$

$\Sigma w = \begin{pmatrix} 0.6 \times 0.0225 + 0.4 \times 0.01125 \\ 0.6 \times 0.01125 + 0.4 \times 0.0625 \end{pmatrix} = \begin{pmatrix} 0.0180 \\ 0.03175 \end{pmatrix}$

MCTR: $[0.0180 / 0.1533, \ 0.03175 / 0.1533] = [11.74\%, \ 20.71\%]$

CCTR: $[0.6 \times 11.74\%, \ 0.4 \times 20.71\%] = [7.04\%, \ 8.28\%]$

Sum of CCTRs: $7.04\% + 8.28\% = 15.33\%$ = $\sigma_p$ (as expected).

Let's now run the simulation and risk decomposition.

```python
# Simulate factor model
n_assets = 50
T = 120
sigma_m = 0.04  # monthly market vol
betas = rng.uniform(0.6, 1.4, n_assets)
sigma_eps = rng.uniform(0.03, 0.08, n_assets)

r_market = rng.normal(0.008, sigma_m, T)
returns_all = np.zeros((T, n_assets))
for i in range(n_assets):
    returns_all[:, i] = betas[i] * r_market + rng.normal(0, sigma_eps[i], T)

# Portfolio vol as function of number of equally-weighted assets
n_range = range(1, n_assets + 1)
port_vols_sim = []
systematic_vols = []
for n in n_range:
    w = np.ones(n) / n
    cov = np.cov(returns_all[:, :n].T)
    port_var = w @ cov @ w
    port_vols_sim.append(np.sqrt(port_var))
    # Systematic component
    port_beta = np.mean(betas[:n])
    systematic_vols.append(port_beta * sigma_m)

fig, ax = plt.subplots()
ax.plot(list(n_range), np.array(port_vols_sim) * np.sqrt(12) * 100, color=PRIMARY, linewidth=2, label='Total risk')
ax.plot(list(n_range), np.array(systematic_vols) * np.sqrt(12) * 100, '--', color=SECONDARY, linewidth=2, label='Systematic risk')
ax.fill_between(list(n_range), np.array(systematic_vols) * np.sqrt(12) * 100, 
                np.array(port_vols_sim) * np.sqrt(12) * 100, alpha=0.2, color=TERTIARY, label='Idiosyncratic (diversifiable)')
ax.set_xlabel('Number of Assets')
ax.set_ylabel('Annualised Volatility (%)')
ax.set_title('Risk Decomposition with Diversification')
ax.legend()
plt.tight_layout()
plt.show()
```

### Interpreting the Risk Decomposition Chart

The chart above is one of the most important diagrams in portfolio theory. Let's read it carefully:

- **Blue solid line (Total risk):** This is the actual annualised volatility of the equal-weight portfolio as we add assets from 1 to 50. It starts high (holding a single stock is very risky) and decreases as we diversify.

- **Coral dashed line (Systematic risk):** This is the component of risk attributable to the market factor. Notice it stays roughly flat (or moves slowly) as we add assets. This is because the average beta of the portfolio converges, but the market risk itself cannot be diversified away.

- **Green shaded area (Idiosyncratic risk):** This is the gap between total risk and systematic risk. It represents the firm-specific risk that is being diversified away. Notice how it shrinks rapidly as $N$ increases --- by the time we have 20-30 assets, most of the idiosyncratic risk has been eliminated.

> **Key Concept:** The green area is the "free lunch" --- risk you can eliminate simply by holding more assets. The coral dashed line is the "price of admission" --- the systematic risk that remains no matter how many assets you hold. Investors are compensated (via higher expected returns) only for bearing systematic risk.

> **CFA Exam Tip:** The CFA curriculum distinguishes between **total risk** ($\sigma_i$), **systematic risk** ($\beta_i \sigma_M$), and **unsystematic risk** ($\sigma_{\epsilon_i}$). For a well-diversified portfolio, total risk $\approx$ systematic risk. This is why CAPM uses $\beta$ (not $\sigma$) to price assets --- in equilibrium, only systematic risk matters.

## 6. Marginal Contribution to Risk

Now we move from theory to practice. A portfolio manager doesn't just want to know that diversification helps --- she needs to know **exactly how much risk each position contributes** to the portfolio.

### Recap of the MCTR Framework

The **marginal contribution to risk (MCTR)** of asset $i$ is:

$$\text{MCTR}_i = \frac{\partial \sigma_P}{\partial w_i} = \frac{(\boldsymbol{\Sigma} \mathbf{w})_i}{\sigma_P}$$

The **component contribution to risk (CCTR)**: $\text{CCTR}_i = w_i \times \text{MCTR}_i$

Key property: $\sum_i \text{CCTR}_i = \sigma_P$ (Euler decomposition).

**Why this matters in practice:**

Imagine you manage a 5-asset portfolio. Each asset has 20% of the capital. But that does NOT mean each asset contributes 20% of the risk! A high-volatility, high-correlation asset might contribute 40% of portfolio risk even though it only has 20% of the capital. Conversely, a low-correlation diversifier might contribute just 5% of risk.

Understanding risk contributions helps you:
- Identify hidden risk concentrations
- Decide which positions to trim (or add) to improve risk efficiency
- Build risk-balanced portfolios (Risk Parity / ERC)

> **Common Mistake:** Don't confuse capital allocation with risk allocation. An equal-weight portfolio (equal capital allocation) almost never has equal risk contributions. Assets with higher volatility and higher correlation to other holdings will dominate the risk budget.

### The Diversification Ratio

How can we measure "how diversified" a portfolio actually is? One elegant metric is the **diversification ratio**:

$$\text{DR} = \frac{\sum_i w_i \sigma_i}{\sigma_p} = \frac{\text{weighted average of individual vols}}{\text{portfolio vol}}$$

**Variable definitions:**

| Symbol | Meaning |
|--------|---------|
| $w_i$ | Weight of asset $i$ in the portfolio |
| $\sigma_i$ | Standard deviation of asset $i$'s returns |
| $\sigma_p$ | Standard deviation of the portfolio's returns |

**In plain English:** The numerator is what portfolio risk *would be* if all assets were perfectly correlated ($\rho = 1$ for every pair). The denominator is what portfolio risk *actually is*. The ratio tells you how much you're benefiting from imperfect correlations.

**Interpretation:**
- If $\text{DR} = 1$: No diversification benefit. All assets are perfectly correlated, or you hold only one asset.
- If $\text{DR} > 1$: The portfolio benefits from diversification. The higher the ratio, the more diversified.
- Typical values for a well-diversified multi-asset portfolio: 1.2 to 1.8.

**Worked example:** Suppose a 50/50 portfolio of two assets with $\sigma_1 = 15\%$, $\sigma_2 = 25\%$, and $\rho = 0.3$.

- Numerator (weighted avg vol): $0.5 \times 0.15 + 0.5 \times 0.25 = 0.20 = 20\%$
- Portfolio variance: $0.5^2(0.0225) + 0.5^2(0.0625) + 2(0.5)(0.5)(0.3)(0.15)(0.25) = 0.005625 + 0.015625 + 0.01125 = 0.0325$
- Portfolio vol: $\sqrt{0.0325} = 18.03\%$
- DR = $20\% / 18.03\% = 1.109$

A DR of 1.109 means the portfolio is about 11% "better" than a perfectly correlated portfolio in terms of risk.

> **Key Concept:** The **Maximum Diversification Portfolio** is the portfolio that maximises the diversification ratio. It was introduced by Choueifaty and Coignard (2008) and is an alternative to mean-variance optimisation that does not require expected return estimates --- only volatilities and correlations.

### Building a 5-Asset Example

Let's compute risk contributions and the diversification ratio for a realistic 5-asset portfolio spanning US Equities, International Equities, Bonds, Real Estate, and Gold. We will first look at the equal-weight portfolio and then compare it to more sophisticated weighting schemes.

The code below defines the asset universe and computes MCTR, CCTR, and percentage risk contributions for an equal-weight portfolio.

```python
# Example with 5 assets
N = 5
asset_names = ['US Equity', 'Intl Equity', 'Bonds', 'Real Estate', 'Gold']
mu_5 = np.array([0.10, 0.12, 0.04, 0.08, 0.06])
sigma_5 = np.array([0.16, 0.22, 0.05, 0.14, 0.18])
corr_5 = np.array([
    [1.0, 0.7, 0.1, 0.5, 0.0],
    [0.7, 1.0, 0.0, 0.4, 0.1],
    [0.1, 0.0, 1.0, 0.2, 0.2],
    [0.5, 0.4, 0.2, 1.0, 0.1],
    [0.0, 0.1, 0.2, 0.1, 1.0],
])
Sigma_5 = np.diag(sigma_5) @ corr_5 @ np.diag(sigma_5)

def risk_contributions(w, Sigma):
    """Compute MCTR, CCTR, and percentage contributions."""
    port_vol = np.sqrt(w @ Sigma @ w)
    mctr = (Sigma @ w) / port_vol
    cctr = w * mctr
    pct_contrib = cctr / port_vol
    return mctr, cctr, pct_contrib, port_vol

# Equal-weight portfolio
w_eq = np.ones(N) / N
mctr, cctr, pct, port_vol = risk_contributions(w_eq, Sigma_5)

print(f"Portfolio volatility: {port_vol*100:.2f}%")
print(f"\n{'Asset':<14} {'Weight':>8} {'MCTR':>8} {'CCTR':>8} {'% Risk':>8}")
print('-' * 50)
for i in range(N):
    print(f"{asset_names[i]:<14} {w_eq[i]:>7.1%} {mctr[i]*100:>7.2f}% {cctr[i]*100:>7.2f}% {pct[i]:>7.1%}")
print(f"{'Total':<14} {'100%':>8} {'':>8} {sum(cctr)*100:>7.2f}% {'100%':>8}")
```

### Interpreting the Risk Contributions

The output above reveals a crucial insight: **capital allocation and risk allocation are very different things.**

Even though each asset has an equal 20% weight, the risk contributions are far from equal:

- **International Equity** likely dominates the risk budget because it has the highest volatility (22%) *and* high correlation with US Equity (0.7). High vol + high correlation = large risk contribution.

- **Bonds** contribute very little risk despite having 20% of the capital. Their low volatility (5%) and low correlations with equities make them excellent diversifiers.

- **Gold** has high individual volatility (18%) but near-zero correlation with US Equity. This low correlation offsets its high volatility, resulting in a moderate risk contribution.

Notice that the CCTRs sum exactly to the portfolio volatility --- this is Euler's decomposition in action.

> **Key Concept:** The risk contribution of an asset depends on three things: (1) its weight, (2) its own volatility, and (3) its correlation with the rest of the portfolio. You cannot assess an asset's risk contribution by looking at any one of these factors in isolation.

> **CFA Exam Tip:** The concept of marginal contribution to risk is directly tested in the CFA Level III curriculum under "Asset Allocation" and "Risk Management." Be prepared to calculate MCTR and CCTR given a covariance matrix and weight vector.

### Capital Allocation vs Risk Allocation

The following pie charts visualise the gap between how capital is distributed (equal across all five assets) and how risk is distributed (dominated by the high-vol, high-correlation assets).

This visual makes a powerful point: **a portfolio that looks balanced in dollar terms may be highly concentrated in risk terms.** Many institutional investors made this mistake before the 2008 financial crisis --- their portfolios looked diversified across asset classes, but 90%+ of their risk came from equities.

```python
# Visualise risk contribution
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

colors_pie = [PRIMARY, SECONDARY, TERTIARY, ACCENT, 'mediumpurple']
ax1.pie(w_eq, labels=asset_names, autopct='%.0f%%', colors=colors_pie, startangle=90)
ax1.set_title('Capital Allocation')

ax2.pie(pct, labels=asset_names, autopct='%.1f%%', colors=colors_pie, startangle=90)
ax2.set_title('Risk Contribution')

plt.suptitle('Equal Weight Portfolio: Capital vs Risk Allocation', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

### Reading the Pie Charts

The contrast between the two pies makes the point vividly:

- **Left pie (Capital Allocation):** Each slice is exactly 20%. The portfolio looks perfectly balanced in terms of dollars invested.

- **Right pie (Risk Contribution):** The slices are highly unequal. The equity assets (US Equity and International Equity) together account for a disproportionately large share of risk, while Bonds contribute a tiny sliver.

This mismatch is the fundamental motivation for **risk parity** (Equal Risk Contribution) portfolios, which we construct next. The idea is: instead of equalising *dollar* weights, equalise *risk* contributions. This typically means holding much more bonds and much less equity than a traditional 60/40 portfolio.

> **Common Mistake:** Many investors think a "60/40 stock/bond" portfolio is well diversified. In reality, about 90% of the risk in a 60/40 portfolio comes from the equity allocation. The 40% in bonds contributes very little risk reduction. True diversification requires thinking in terms of *risk allocation*, not *capital allocation*.

## 7. Risk Budgeting --- Equal Risk Contribution (ERC) Portfolio

### The Problem

We just saw that an equal-weight portfolio has highly unequal risk contributions. Can we find weights such that every asset contributes the **same amount of risk** to the portfolio?

### The ERC Condition

The **ERC portfolio** (also called the **risk parity** portfolio) equalises each asset's percentage risk contribution:

$$w_i \cdot \text{MCTR}_i = w_j \cdot \text{MCTR}_j \quad \forall \, i, j$$

In other words: $\text{CCTR}_1 = \text{CCTR}_2 = \ldots = \text{CCTR}_N = \frac{\sigma_p}{N}$

**In plain English:** Every asset contributes exactly $1/N$ of the portfolio's total risk. If you have 5 assets, each contributes exactly 20% of the risk.

### How to Find ERC Weights

There is generally no closed-form solution for ERC weights (except in special cases like when all assets have the same volatility). We must solve an optimisation problem.

One common formulation minimises the sum of squared differences between actual and target risk contributions:

$$\min_{\mathbf{w}} \sum_{i=1}^{N} \sum_{j=1}^{N} \left(w_i (\boldsymbol{\Sigma}\mathbf{w})_i - w_j (\boldsymbol{\Sigma}\mathbf{w})_j\right)^2$$

Subject to: $\sum_i w_i = 1$ and $w_i > 0$.

**What to expect:** Since bonds have much lower volatility than equities, the ERC portfolio will have a much larger bond allocation (to make bonds contribute as much risk as equities). This is the hallmark of risk parity strategies.

> **Key Concept:** Risk parity gained fame after Ray Dalio's Bridgewater Associates popularised the "All Weather" portfolio, which uses risk parity principles. The idea is that by balancing risk across asset classes (rather than balancing capital), you create a portfolio that performs reasonably well in all economic environments --- inflation, deflation, growth, and recession.

> **CFA Exam Tip:** The CFA Level III curriculum covers risk parity / risk budgeting under Asset Allocation. You should understand: (1) the difference between equal weight and equal risk contribution, (2) that risk parity typically involves leveraging bonds to make their risk contribution comparable to equities, and (3) the advantages (true diversification) and disadvantages (requires leverage, sensitive to correlation estimates) of risk parity.

Let's compute the ERC portfolio for our 5-asset example.

### Monte Carlo Portfolio Simulation

Finally, let's throw thousands of random portfolios at the wall and see what sticks. By generating random weight combinations, we can visualise the **feasible set** --- all possible portfolios --- and identify the **efficient frontier**.

**What is the feasible set?** It is the set of all possible risk-return combinations achievable by varying the portfolio weights across our 5 assets (with the constraint that weights are non-negative and sum to 1).

**What is the efficient frontier?** It is the upper-left boundary of the feasible set --- the set of portfolios that offer the maximum expected return for each level of risk (or equivalently, the minimum risk for each level of expected return). Every rational investor should hold a portfolio on the efficient frontier.

> **What to watch for:** The random portfolios will form a "bullet" shape (sometimes called the Markowitz bullet). The left edge of this bullet is the efficient frontier --- portfolios that offer the maximum return for each level of risk.

> **Key Concept:** Any portfolio that lies *inside* the feasible set (not on the efficient frontier) is **dominated** --- there exists another portfolio with either higher return for the same risk, or lower risk for the same return. Rational investors should never hold a dominated portfolio.

> **CFA Exam Tip:** The efficient frontier is a central concept tested across all three CFA levels. At Level I, you need to understand what it represents and how diversification creates it. At Level II, you compute it. At Level III, you use it for asset allocation decisions.

```python
def erc_portfolio(Sigma, budget=None):
    """Equal Risk Contribution portfolio via optimisation."""
    n = Sigma.shape[0]
    if budget is None:
        budget = np.ones(n) / n  # equal risk budget
    
    def objective(w):
        port_vol = np.sqrt(w @ Sigma @ w)
        mctr = (Sigma @ w) / port_vol
        cctr = w * mctr
        target = budget * port_vol
        return np.sum((cctr - target)**2)
    
    constraints = [{'type': 'eq', 'fun': lambda w: np.sum(w) - 1}]
    bounds = [(0.01, None) for _ in range(n)]
    w0 = np.ones(n) / n
    
    result = optimize.minimize(objective, w0, method='SLSQP',
                               bounds=bounds, constraints=constraints,
                               options={'ftol': 1e-15, 'maxiter': 1000})
    return result.x

w_erc = erc_portfolio(Sigma_5)
mctr_erc, cctr_erc, pct_erc, vol_erc = risk_contributions(w_erc, Sigma_5)

print("ERC Portfolio:")
print(f"Portfolio volatility: {vol_erc*100:.2f}%")
print(f"\n{'Asset':<14} {'Weight':>8} {'CCTR':>8} {'% Risk':>8}")
print('-' * 40)
for i in range(N):
    print(f"{asset_names[i]:<14} {w_erc[i]:>7.1%} {cctr_erc[i]*100:>7.2f}% {pct_erc[i]:>7.1%}")

print(f"\nRisk contributions are approximately equal: max deviation = {(pct_erc.max() - pct_erc.min())*100:.4f}%")
```

### Interpreting the ERC Portfolio

Compare the ERC weights with the equal weights:

- **Bonds** receive a much larger weight than 20%. This makes sense: bonds have low volatility (5%), so you need to hold a lot of them to make their risk contribution equal to the equities.

- **International Equity** receives a smaller weight than 20%. It has the highest volatility (22%) and high correlation with other assets, so a small allocation already contributes substantial risk.

- The percentage risk contributions are now approximately equal at 20% each (within numerical precision of the optimiser).

**The trade-off:** The ERC portfolio has lower overall volatility than the equal-weight portfolio (because it tilts toward bonds), but it also has a lower expected return. This is the classic risk parity dilemma --- to achieve the same expected return as a traditional portfolio, risk parity strategies often need to use **leverage** (borrow money to scale up the low-risk portfolio).

> **Common Mistake:** Risk parity is sometimes marketed as "free lunch 2.0" --- better diversification with no cost. In reality, risk parity portfolios often have lower expected returns than traditional portfolios (because they hold more bonds). The leverage required to match traditional returns introduces its own risks (margin calls, borrowing costs, liquidity risk).

> **CFA Exam Tip:** If asked about the advantages of risk parity, cite: true risk diversification, no reliance on expected return estimates. If asked about disadvantages, cite: requires leverage, sensitive to correlation estimates, may underperform in strong equity markets.

### Interpreting the Monte Carlo Cloud

The scatter plot of random portfolios reveals the **feasible set** --- every possible portfolio you could construct from these assets. The key observations:

1. **No portfolios exist in the upper-left region** --- you can't get high return with low risk (if markets are efficient)
2. **The left boundary** is the efficient frontier --- the best risk-return trade-offs available
3. **Most random portfolios are inefficient** --- they sit well inside the frontier
4. **The minimum variance portfolio** is the leftmost point on the frontier

> **Practical Insight:** This is why portfolio optimisation matters. Simply picking random weights almost never gives you an efficient portfolio. Even naive diversification (equal weights) usually beats random weights.

## 8. Diversification Ratio

Let's now compute the diversification ratio for our three portfolio strategies: equal weight, ERC, and the maximum diversification portfolio.

### Formal Definition

The **diversification ratio** measures the benefit of diversification:

$$\text{DR} = \frac{\mathbf{w}^\top \boldsymbol{\sigma}}{\sqrt{\mathbf{w}^\top \boldsymbol{\Sigma} \mathbf{w}}} = \frac{\text{Weighted avg volatility}}{\text{Portfolio volatility}}$$

**In plain English:** The numerator answers "what would portfolio risk be if all assets moved in perfect lockstep?" The denominator is the actual portfolio risk. The ratio is always $\geq 1$, with equality only when all correlations are 1 (no diversification).

**Why DR $\geq 1$ always holds:** By Cauchy-Schwarz (or simply because correlations $\leq 1$ reduce portfolio variance below the weighted-average-volatility squared):

$$\sigma_p = \sqrt{\mathbf{w}^\top \Sigma \mathbf{w}} \leq \mathbf{w}^\top \boldsymbol{\sigma}$$

with equality iff $\rho_{ij} = 1$ for all $i, j$. So $\text{DR} \geq 1$ always.

### Three Portfolios to Compare

We will compute the DR for:

1. **Equal Weight** --- the simplest approach (all weights = 20%)
2. **Equal Risk Contribution (ERC)** --- equalises risk contributions (risk parity)
3. **Maximum Diversification** --- explicitly maximises the diversification ratio

The Maximum Diversification Portfolio solves:

$$\max_{\mathbf{w}} \frac{\mathbf{w}^\top \boldsymbol{\sigma}}{\sqrt{\mathbf{w}^\top \boldsymbol{\Sigma} \mathbf{w}}} \quad \text{s.t.} \quad \mathbf{w}^\top \mathbf{1} = 1, \; w_i \geq 0$$

This is equivalent to finding the portfolio with the maximum Sharpe ratio when all assets have the same expected return (equal to their volatility). It is a pure "diversification-seeking" strategy.

Let's compute all three and compare.

```python
def diversification_ratio(w, sigma_vec, Sigma):
    weighted_avg_vol = w @ sigma_vec
    port_vol = np.sqrt(w @ Sigma @ w)
    return weighted_avg_vol / port_vol

dr_eq = diversification_ratio(w_eq, sigma_5, Sigma_5)
dr_erc = diversification_ratio(w_erc, sigma_5, Sigma_5)

print(f"Diversification ratio (equal weight):  {dr_eq:.4f}")
print(f"Diversification ratio (ERC):           {dr_erc:.4f}")

# Maximum diversification portfolio
def max_div_portfolio(sigma_vec, Sigma):
    n = len(sigma_vec)
    def neg_dr(w):
        return -(w @ sigma_vec) / np.sqrt(w @ Sigma @ w)
    constraints = [{'type': 'eq', 'fun': lambda w: np.sum(w) - 1}]
    bounds = [(0, 1) for _ in range(n)]
    w0 = np.ones(n) / n
    result = optimize.minimize(neg_dr, w0, method='SLSQP', bounds=bounds, constraints=constraints)
    return result.x

w_md = max_div_portfolio(sigma_5, Sigma_5)
dr_md = diversification_ratio(w_md, sigma_5, Sigma_5)
print(f"Diversification ratio (max div port):  {dr_md:.4f}")

print(f"\nMax Div Portfolio weights:")
for name, wi in zip(asset_names, w_md):
    print(f"  {name:<14} {wi:.1%}")
```

### Interpreting the Diversification Ratios

The output shows three diversification ratios, one for each strategy:

- **Equal Weight (DR $\approx$ 1.2--1.4):** Decent diversification, but not optimal. The equal-weight portfolio doesn't account for differences in volatility or correlation.

- **ERC (slightly higher DR):** By equalising risk contributions, the ERC portfolio achieves somewhat better diversification. It naturally tilts toward low-correlation assets.

- **Max Diversification (highest DR):** By construction, this portfolio has the highest possible diversification ratio. It typically tilts heavily toward low-correlation assets --- you'll likely see large weights on Gold and Bonds, which have near-zero correlation with equities.

**Look at the Max Div Portfolio weights:** Notice how it loads up on the assets with the lowest correlations to everything else. Bonds (low vol, low correlation) and Gold (moderate vol, near-zero correlation with equities) are typically favoured. International Equity, despite its high expected return, may get a low weight because it's so correlated with US Equity.

> **Key Concept:** The Maximum Diversification Portfolio is a powerful concept because it requires NO expected return inputs --- only the covariance matrix. This sidesteps the well-known problem that expected return estimates are notoriously unreliable and cause mean-variance optimisation to produce unstable, extreme portfolios.

> **CFA Exam Tip:** When comparing portfolio construction approaches, remember the spectrum: Equal Weight (simplest, no inputs needed) $\to$ ERC / Risk Parity (needs covariance matrix) $\to$ Max Diversification (needs covariance matrix) $\to$ Mean-Variance Optimisation (needs covariance matrix AND expected returns). Each step adds complexity and potential for estimation error.

## 9. International Diversification

One of the classic arguments for diversification is to invest across countries. Different economies have different business cycles, monetary policies, and structural characteristics, so international stocks should have lower correlations than domestic stocks.

### The Case for International Diversification

**Why it should work:**
- Different countries have different economic drivers (US is tech-heavy, Brazil is commodity-heavy, Japan is manufacturing-heavy)
- Currency movements add another source of return variation
- Political and regulatory environments differ
- Business cycles are not perfectly synchronised

**The correlation evidence:** Cross-country equity correlations are typically in the range of 0.3--0.7, lower than within-country correlations (which are often 0.5--0.9). Emerging markets tend to have lower correlations with developed markets, offering greater diversification potential.

Adding international assets increases diversification because cross-country correlations are typically lower than within-country correlations.

### The Problem: Correlation Breakdown

**But does it still work when you need it most?** Research shows that correlations between markets **increase during crises** --- exactly when you need diversification the most. This is called **correlation breakdown** or **contagion**.

During the 2008 Global Financial Crisis, correlations between major equity markets spiked above 0.90. During the COVID-19 crash of March 2020, the same thing happened. When panic hits, "all correlations go to 1" --- investors sell everything simultaneously, and the diversification benefit evaporates precisely when it is most needed.

> **Common Mistake:** Don't assume that historical correlations during calm periods will hold during crashes. Tail correlations are almost always higher than normal-period correlations. This means that risk models built on normal-period data will *underestimate* portfolio risk during crises.

> **CFA Exam Tip:** The CFA curriculum discusses the "correlation breakdown" phenomenon under the topic of international diversification. Key points to remember: (1) correlations increase in bear markets, (2) this reduces the effectiveness of international diversification precisely when it's most needed, (3) alternative assets (real assets, gold, Treasury bonds) may provide better tail-risk diversification than international equities.

### Visualising Cross-Country Correlations

### Building International Portfolios Step by Step

Now let's see what happens as we sequentially add countries to an equal-weight portfolio. We start with US-only and progressively add UK, Germany, Japan, China, Brazil, and India.

**What to expect:** Each additional country should reduce portfolio volatility (more diversification), but with diminishing benefits. The biggest improvement should come from adding the first few countries, especially those with low correlation to the existing portfolio.

```python
# Benefit of adding countries incrementally
Sigma_intl = np.diag(sigma_intl) @ corr_intl @ np.diag(sigma_intl)

port_vols_intl = []
port_rets_intl = []
for n in range(1, n_c + 1):
    w = np.ones(n) / n
    S = Sigma_intl[:n, :n]
    port_vols_intl.append(np.sqrt(w @ S @ w))
    port_rets_intl.append(w @ mu_intl[:n])

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.bar(range(1, n_c + 1), np.array(port_vols_intl) * 100, color=PRIMARY, edgecolor='white')
ax1.set_xticks(range(1, n_c + 1))
ax1.set_xticklabels(['+'.join(countries[:i]) if i <= 3 else f'{i} countries' for i in range(1, n_c + 1)], 
                     rotation=45, ha='right', fontsize=8)
ax1.set_ylabel('Portfolio Volatility (%)')
ax1.set_title('Diversification Across Countries')

ax2.scatter(np.array(port_vols_intl) * 100, np.array(port_rets_intl) * 100, 
            s=100, color=PRIMARY, edgecolors='black', zorder=5)
for i, c in enumerate(countries):
    ax2.plot(sigma_intl[i] * 100, mu_intl[i] * 100, 'o', color=SECONDARY, markersize=8)
    ax2.annotate(c, (sigma_intl[i] * 100 + 0.3, mu_intl[i] * 100), fontsize=8)
for i in range(len(port_vols_intl)):
    ax2.annotate(f'{i+1}', (port_vols_intl[i] * 100 - 0.3, port_rets_intl[i] * 100 + 0.2), fontsize=9, fontweight='bold')
ax2.set_xlabel('Volatility (%)')
ax2.set_ylabel('Expected Return (%)')
ax2.set_title('International EW Portfolios in Risk-Return Space')

plt.tight_layout()
plt.show()
```

### Interpreting the International Diversification Charts

**Left panel (Bar chart --- Volatility vs Number of Countries):**

- Starting with the US alone (16% vol), adding the UK brings a meaningful reduction. Adding Germany helps further.
- The biggest drops typically occur when adding countries with *low* correlation to the existing portfolio. Adding Japan (lower correlation with the US/UK/Germany block) provides a notable benefit.
- Adding all 7 countries brings the portfolio volatility significantly below any individual country's volatility. This is the international diversification benefit in action.

**Right panel (Risk-Return scatter):**

- The **coral dots** are individual countries. Brazil has the highest volatility (30%) and highest expected return (14%). Japan has low expected return (6%) but moderate volatility (19%).
- The **blue dots** (numbered 1--7) are the equal-weight portfolios with increasing numbers of countries. Notice how they move to the *left* of individual country dots --- this is the diversification benefit creating portfolios that are less risky than any individual component.
- The progression from point 1 to point 7 shows the risk-return trade-off of adding countries. Each point represents a different level of international diversification.

> **Key Concept:** International diversification allows you to access a set of risk-return combinations that are impossible with domestic-only investing. The equal-weight international portfolio has lower risk than *any* individual country --- this is the power of combining imperfectly correlated return streams.

> **Common Mistake:** The order in which we add countries matters for this incremental analysis. We chose an arbitrary order (US first, then UK, etc.). In practice, you would use optimisation to determine the best *combination* of countries, not just add them sequentially.

> **CFA Exam Tip:** The CFA curriculum emphasises that the benefits of international diversification have been *declining* over time as markets become more integrated (globalisation). Correlations between developed markets have increased from roughly 0.3-0.4 in the 1980s to 0.6-0.8 today. The greatest remaining diversification benefits come from emerging and frontier markets.

## 10. Monte Carlo Portfolio Simulation

We now bring everything together with a Monte Carlo simulation. The idea is beautifully simple:

1. Generate thousands of **random portfolio weight vectors** (using a Dirichlet distribution to ensure weights are non-negative and sum to 1)
2. For each random portfolio, compute its **expected return**, **volatility**, and **Sharpe ratio**
3. Plot all portfolios in **risk-return space** to visualise the feasible set
4. Overlay the **efficient frontier** computed via optimisation

### Why Monte Carlo?

With 5 assets, the set of all possible portfolios is a 4-dimensional object (one weight is determined by the other four summing to 1). We can't visualise 4D directly, but we *can* project every portfolio onto the 2D risk-return plane. By generating enough random portfolios, we "fill in" this projection and reveal the shape of the feasible set.

### The Dirichlet Distribution for Random Weights

We use the Dirichlet distribution with all concentration parameters equal to 1: $\mathbf{w} \sim \text{Dir}(1, 1, \ldots, 1)$. This generates uniformly distributed weight vectors on the simplex $\{\mathbf{w} : w_i \geq 0, \sum w_i = 1\}$. It's the natural distribution for "pick random portfolio weights."

### The Efficient Frontier

To trace the efficient frontier, we solve a series of optimisation problems: for each target return level $\mu^*$, find the portfolio with minimum variance:

$$\min_{\mathbf{w}} \mathbf{w}^\top \Sigma \mathbf{w} \quad \text{s.t.} \quad \mathbf{w}^\top \boldsymbol{\mu} = \mu^*, \; \mathbf{w}^\top \mathbf{1} = 1, \; w_i \geq 0$$

Varying $\mu^*$ across the range of achievable returns traces out the efficient frontier.

Generate thousands of random portfolios to visualise the feasible region and identify the efficient frontier empirically.

Let's run the simulation with 15,000 random portfolios for our 5-asset universe.

```python
def random_portfolios(mu, Sigma, n_portfolios=10000):
    """Generate random long-only portfolio weights and compute risk-return."""
    n = len(mu)
    results = np.zeros((n_portfolios, 3))  # return, vol, sharpe
    weights_record = np.zeros((n_portfolios, n))
    
    for i in range(n_portfolios):
        # Random weights via Dirichlet distribution
        w = rng.dirichlet(np.ones(n))
        ret = w @ mu
        vol = np.sqrt(w @ Sigma @ w)
        sr = (ret - 0.03) / vol
        results[i] = [ret, vol, sr]
        weights_record[i] = w
    
    return results, weights_record

results, all_weights = random_portfolios(mu_5, Sigma_5, 15000)

# Find efficient frontier numerically
def min_var_for_target(mu, Sigma, target):
    n = len(mu)
    def obj(w): return w @ Sigma @ w
    cons = [{'type': 'eq', 'fun': lambda w: np.sum(w) - 1},
            {'type': 'eq', 'fun': lambda w: w @ mu - target}]
    bounds = [(0, 1)] * n
    res = optimize.minimize(obj, np.ones(n)/n, method='SLSQP', bounds=bounds, constraints=cons)
    return np.sqrt(res.fun), res.x

targets = np.linspace(mu_5.min() + 0.005, mu_5.max() - 0.005, 50)
ef_pts = [min_var_for_target(mu_5, Sigma_5, t) for t in targets]
ef_vols = [p[0] for p in ef_pts]
ef_rets = targets

fig, ax = plt.subplots(figsize=(10, 7))
scatter = ax.scatter(results[:, 1] * 100, results[:, 0] * 100, c=results[:, 2], 
                     cmap='viridis', s=5, alpha=0.5)
ax.plot(np.array(ef_vols) * 100, ef_rets * 100, color=SECONDARY, linewidth=3, label='Efficient Frontier')

# Mark individual assets
for i in range(N):
    ax.plot(sigma_5[i] * 100, mu_5[i] * 100, 'D', color='white', markersize=10, 
            markeredgecolor='black', markeredgewidth=2, zorder=5)
    ax.annotate(asset_names[i], (sigma_5[i] * 100 + 0.3, mu_5[i] * 100 + 0.2), fontsize=9, fontweight='bold')

plt.colorbar(scatter, label='Sharpe Ratio')
ax.set_xlabel('Portfolio Volatility (%)')
ax.set_ylabel('Expected Return (%)')
ax.set_title('Monte Carlo Simulation \u2014 15,000 Random Portfolios')
ax.legend(fontsize=12)
plt.tight_layout()
plt.show()
```

### Reading the Monte Carlo Scatter Plot

This chart is the grand synthesis of everything we have covered. Let's read it carefully:

**The cloud of points:** Each tiny dot represents one of 15,000 random portfolios. The colour indicates the Sharpe ratio (risk-adjusted return) --- brighter (yellow) means higher Sharpe, darker (purple) means lower Sharpe.

**The Markowitz bullet:** Notice the characteristic "bullet" or "egg" shape of the feasible set. This shape is a fundamental result of portfolio theory --- it arises from the mathematics of combining assets with imperfect correlations.

**The efficient frontier (coral line):** This is the upper-left boundary of the feasible set. Portfolios on this line offer the best possible return for each level of risk. Key observations:
- The efficient frontier curves to the left of every individual asset. This means that *every* optimal portfolio has lower risk than some (or all) of its component assets. This is diversification in action.
- The highest Sharpe ratio (the tangency portfolio) is the point on the frontier where a line from the risk-free rate (3%) is tangent to the frontier.

**Individual assets (diamond markers):** Notice where each asset sits relative to the efficient frontier:
- All individual assets lie *inside* (to the right of) the efficient frontier. This means holding any single asset is suboptimal --- you can always do better by diversifying.
- Bonds have low risk and low return (bottom-left). Equities have high risk and high return (upper-right). Gold is somewhere in between.

**Most random portfolios are inefficient:** The vast majority of dots lie well inside the frontier. This drives home the point that *random* diversification is better than holding a single asset, but far worse than *optimal* diversification.

> **Key Concept:** The efficient frontier is the central object of modern portfolio theory (MPT), introduced by Harry Markowitz in 1952. It demonstrates that the risk-return trade-off is not a straight line --- diversification "bends" the frontier to the left, creating opportunities that no individual asset offers.

> **CFA Exam Tip:** Be prepared to identify the efficient frontier, the minimum variance portfolio, and the global minimum variance portfolio on a chart. Know that: (1) portfolios below the minimum variance portfolio on the frontier are dominated, (2) the tangency portfolio maximises the Sharpe ratio, (3) with a risk-free asset, the efficient frontier becomes the **Capital Market Line (CML)**, a straight line from the risk-free rate through the tangency portfolio.

> **Common Mistake:** Mean-variance optimisation (the math behind the efficient frontier) is elegant but fragile in practice. Small changes in expected return estimates can cause large swings in optimal weights. This is why practitioners often prefer more robust approaches like equal weight, risk parity, or maximum diversification --- strategies that don't require expected return estimates.

## 11. Summary and Key Takeaways

Let's recap the most important ideas from this notebook:

### The Big Ideas

1. **Diversification is the only free lunch in finance.** Combining assets with $\rho < 1$ reduces portfolio risk without reducing expected return. This is a mathematical certainty, not a conjecture.

2. **Diversification has limits.** You cannot diversify away systematic (market) risk. For an equal-weight portfolio, the irreducible risk floor is $\sigma\sqrt{\rho}$.

3. **Correlation is the key variable.** Lower correlation = more diversification benefit. At $\rho = -1$, risk can be completely eliminated. At $\rho = +1$, there is no benefit at all.

4. **Capital allocation $\neq$ risk allocation.** An equal-weight portfolio can have highly unequal risk contributions. Understanding MCTR and CCTR is essential for risk management.

5. **Risk parity (ERC)** equalises risk contributions rather than capital weights. It requires more bonds and less equity than traditional portfolios, but achieves true risk diversification.

6. **The diversification ratio** is a clean, single-number measure of portfolio diversification. The Maximum Diversification Portfolio maximises it without requiring expected return estimates.

7. **International diversification** helps but has limitations: correlations increase during crises (when diversification is most needed), and globalisation has been increasing cross-country correlations over time.

8. **The efficient frontier** is the set of portfolios offering the best risk-return trade-off. Most randomly constructed portfolios are far from efficient.

### Formulas to Memorise

| Formula | Use Case |
|---------|----------|
| $\sigma_p^2 = w_1^2\sigma_1^2 + w_2^2\sigma_2^2 + 2w_1w_2\rho\sigma_1\sigma_2$ | Two-asset portfolio variance |
| $\sigma_p^2 = \frac{\sigma^2}{N} + \frac{N-1}{N}\rho\sigma^2$ | Equal-weight, identical assets |
| $\text{MCTR}_i = \frac{(\Sigma\mathbf{w})_i}{\sigma_p}$ | Marginal contribution to risk |
| $\text{DR} = \frac{\mathbf{w}^\top\boldsymbol{\sigma}}{\sigma_p}$ | Diversification ratio |
| $w_1^* = \frac{\sigma_2}{\sigma_1 + \sigma_2}$ | Zero-risk weight when $\rho = -1$ |
